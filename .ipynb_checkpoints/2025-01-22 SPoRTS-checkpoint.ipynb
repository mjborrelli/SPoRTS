```python
def celltype_init(celltype, report_info = True):
    # checks if input files match any entries in biosensor_settings.
    # if so, uses the indicated settings; if not, sets to defaults.
    # then, checks if files have been provided in pairs and
    # identifies any unpaired images

    global biosensor
    global config
    global num_fluor
    global denom_fluor
    global ratio_expression
    global graphs_cmap
    global rmaps_cmap
    
    try:
        biosensor = celltype.split("-")[1]
        biosensor_info = [item for item in biosensor_settings if item[0] == biosensor][0]
        biosensor_info = check_biosensor_info(biosensor_info, report_info = report_info)
    except:
        if "+" in biosensor:
            biosensor_info = [biosensor, biosensor.split("+")[0], biosensor.split("+")[1], "A/B", default_graphs_cmap, default_rmaps_cmap]
        else:
            print(f'No biosensor information is provided for {celltype}. Attempting to parse settings based on file names...')
            try:
                celltype_fluors = list(sorted(set([y[2] for y in file_details if y[0] == celltype])))
                biosensor_info = [biosensor, celltype_fluors[0], celltype_fluors[1], "A/B", default_graphs_cmap, default_rmaps_cmap]
            except:
                print(f'Could not parse biosensor_settings for {celltype}. Please check file names or switch to intensity analysis.')
                return([])
                
    biosensor = biosensor_info[0]
    config = biosensor_info[3]
    if config == "A/B":
        num_fluor, denom_fluor = biosensor_info[1], biosensor_info[2]
    elif config == "B/A":
        num_fluor, denom_fluor = biosensor_info[2], biosensor_info[1]
    ratio_expression = f'{num_fluor}/{denom_fluor}'
    graphs_cmap = biosensor_info[4]
    rmaps_cmap = biosensor_info[5]
    
    # finds and reports unpaired files if present
    print(f'Checking for unpaired files...')
    celltype_flist = [item for item in file_details if item[0] == celltype]
    num_flist = [x for x in celltype_flist if x[2] == num_fluor]
    denom_flist = [x for x in celltype_flist if x[2] == denom_fluor]
    num_filtered = [x for x in num_flist if (("_".join(x)).replace(f'_{num_fluor}_', f'_{denom_fluor}_')).split("_") in denom_flist]
    denom_filtered = [x for x in denom_flist if (("_".join(x)).replace(f'_{denom_fluor}_', f'_{num_fluor}_')).split("_") in num_flist]
    filtered_celltype_flist = sorted(num_filtered + denom_filtered)
    unpaired_files = sorted([x for x in celltype_flist if x not in filtered_celltype_flist])
    if unpaired_files:
        print(f'\nFound unpaired files for {celltype}; stop SPoRTS and resolve if necessary.\nThe following files are unpaired:')
        for file_name in unpaired_files:
            print("_".join(file_name))
    else:
        print(f'\nNo unpaired files for {celltype}.\n')
    return(filtered_celltype_flist)
```

```python
def check_biosensor_info(biosensor_info, report_info = True):
    # checks that file naming and assigned biosensor_info are valid/
    # compatible. this function only really does anything if the
    # biosensor_info provided is inaccurate or incorrectly formatted.
    # in such a case, this function tries to parse biosensor_info and
    # input files by assigning "likely" identifiers and user is
    # notified of any issues that are encountered in the process
    
    checked_biosensor_info = [biosensor_info[0]]
    if report_info:
        print(f'\nConfirming valid biosensor_settings for {biosensor_info[0]} ({celltype})...')
    error_messages = []
    
    # identify and assign fluorophores based on file naming
    celltype_fluors = list(dict.fromkeys([y[2] for y in file_details if y[0] == celltype]))
    if len(celltype_fluors) < 2:
        raise ValueError()
    if len(biosensor_info) >= 3:
        likely_fluors = list(dict.fromkeys([x for x in (biosensor_info[1], biosensor_info[2]) if x in celltype_fluors]))
    elif len(biosensor_info) == 2:
        likely_fluors = biosensor_info[1]
    elif len(biosensor_info) == 1:
        likely_fluors = []
        
    if len(likely_fluors) == 2:
        likely_fluorA = biosensor_info[1]
        likely_fluorB = biosensor_info[2]
    else:
        error_messages.append(f'\nProblem encountered:')
        possible_fluors = list(dict.fromkeys([x for x in biosensor_info if x in celltype_fluors and not x in likely_fluors]))
        if len(likely_fluors) == 1:
            likely_fluorA = likely_fluors[0]
            if possible_fluors:
                likely_fluorB = [x for x in possible_fluors if x != likely_fluorA][0]
                error_messages.append(f'\tOne fluorophore argument out of place.\n\tfluorA and fluorB assigned by order of appearance.')
            else:
                likely_fluorB = [x for x in celltype_fluors if x != likely_fluorA][0]
                error_messages.append(f'\tOne fluorophore argument missing or mis-matched.\n\tExisting fluorophore assigned to fluorA; fluorB inferred from file names.')
        elif not likely_fluors:
            if len(possible_fluors) >= 2:
                likely_fluorA = possible_fluors[0]
                likely_fluorB = possible_fluors[1]
                error_messages.append(f'\tFluorophore arguments out of place.\n\tfluorA and fluorB assigned by order of appearance.')
            elif possible_fluors:
                likely_fluorA = possible_fluors[0]
                likely_fluorB = [x for x in celltype_fluors if x != likely_fluorA][0]
                error_messages.append(f'\tOne fluorophore argument out of place, one fluorophore argument missing or mis-matched.\n\tExisting fluorophore assigned to fluorA; fluorB inferred from file names.')
            else:
                likely_fluorA = celltype_fluors[0]
                likely_fluorB = celltype_fluors[1]
                error_messages.append(f'\tFluorophore arguments missing.\n\tfluorA and fluorB inferred from file names.')
    checked_biosensor_info.extend([likely_fluorA, likely_fluorB])
    
    # identify and assign numerator/denominator configuration
    try:
        likely_config = biosensor_info[3]
        if not likely_config in ("A/B", "B/A"):
            raise ValueError()
    except:
        error_messages.append(f'\nProblem encountered:')
        likely_config = list(dict.fromkeys([x for x in biosensor_info if x in ("A/B", "B/A")]))
        if len(likely_config) == 1:
            likely_config = likely_config[0]
            error_messages.append(f'\tNumerator/denominator configuration out of place.')
        elif len(likely_config) > 1:
            likely_config = likely_config[0]
            error_messages.append(f'\tNumerator/denominator configuration ambiguous (both "A/B" and "B/A" given).\n\tNumerator/denominator configuration assigned by first appearance.')
        elif not likely_config:
            likely_config = "A/B"
            error_messages.append(f'\tNumerator/denominator configuration missing; assigned to default "{likely_config}".')
    checked_biosensor_info.append(likely_config)
    if likely_config == "A/B":
        likely_ratio_expression = f'{likely_fluorA}/{likely_fluorB}'
    elif likely_config == "B/A":
        likely_ratio_expression = f'{likely_fluorB}/{likely_fluorA}'
        
    # identify and assign colormaps
    likely_cmaps = list(dict.fromkeys([x for x in biosensor_info if ((x in plt.colormaps()) or (x.startswith("cmr.")))]))
    if not likely_cmaps:
        likely_graphs_cmap = default_graphs_cmap
        likely_rmaps_cmap = default_rmaps_cmap
        error_messages.append(f'\nNo colormaps specified; assigning defaults.')
    elif len(likely_cmaps) == 1:
        likely_graphs_cmap = likely_cmaps[0]
        likely_rmaps_cmap = likely_cmaps[0]
        error_messages.append(f'\nOne colormap specified; assigning to both graphs and ratio-mapped images.')
    elif len(likely_cmaps) >= 2:
        if len(likely_cmaps) == 2:
            likely_graphs_cmap = likely_cmaps[0]
            likely_rmaps_cmap = likely_cmaps[1]
        else:
            error_messages.append(f'\nToo many colormaps specified; assigning to graphs and ratio-mapped images by order of appearance.')
    checked_biosensor_info.extend([likely_graphs_cmap, likely_rmaps_cmap])
    
    # report any issues encountered and final configuration. if user
    # finds that the configuration is not as intended, the error
    # messages explain what might be going wrong
    if report_info:
        if error_messages:
            for x in error_messages:
                print(x)
    
        print(f'\nFinal biosensor_settings:\n',
              f'\tbiosensor = {checked_biosensor_info[0]}\n',
              f'\tfluorA = {checked_biosensor_info[1]}\n',
              f'\tfluorB = {checked_biosensor_info[2]}\n',
              f'\tnumerator/denominator configuraton = {likely_ratio_expression}\n',
              f'\tgraphs_cmap = {checked_biosensor_info[4]}\n',
              f'\trmaps_cmap = {checked_biosensor_info[5]}\n')
    return(checked_biosensor_info)
```

```python
def cleanup_df(*dfs):
    # removes initial_col from DataFrame, which was used to set up
    # the index, and removes any rows and columns containing only NaN
    # values. more or less functions like "cropping" to useful data
    
    clean_dfs = []
    for df in dfs:
        try:
            df.drop('initial_col', axis = 1, inplace = True)
        except:
            pass
        try:
            df.dropna(axis = 1, how = 'all', inplace = True)
            df.dropna(axis = 0, how = 'all', inplace = True)
        except:
            pass
        df = df.reset_index(drop = True)
        clean_dfs.append(df)
    if len(clean_dfs) == 1:
        clean_dfs = clean_dfs[0]
    return(clean_dfs)
```

```python
def constitutive_correct_ratiomap(df, ccorr_data):
    # two-dimensional constitutive correction. see paper (Fig. S1F)
    
    # determine foreground coordinates
    foreground_coords = np.argwhere(df > 0)
    
    # calculate centroid (geometric mean) pixel location
    centroid = np.mean(foreground_coords, axis = 0, dtype = int)
    
    # create a DataFrame to be populated with distance values
    dist_df = pd.DataFrame(index = range(df.shape[0]), columns = range(df.shape[1]))
    
    # determine Euclidean distance between each foreground pixel and
    # centroid. populate distance DataFrame with these values
    for coord in foreground_coords:
        dist_df.iloc[coord[0], coord[1]] = math.dist([coord[0], coord[1]], centroid)
        
    # convert distance DataFrame to relative distance dataframe.
    # accounts for small numbers of pixels that are outside the main
    # foreground (i.e., loose cells, etc.)
    rel_dist_df = (dist_df / np.percentile(dist_df.values[dist_df.notna()], 99)).clip(upper = 1)
    
    # create a DataFrame to be populated with constitutive correction
    # values
    correction_df = pd.DataFrame(index = range(df.shape[0]), columns = range(df.shape[1]))
    
    # determine the value from the constitutive correction data that
    # corresponds to the relative distance for a given pixel to the
    # centroid and populate the correction DataFrame using these
    # values
    for coord in foreground_coords:
        correction_df.iloc[coord[0], coord[1]] = ccorr_data[int(rel_dist_df.iloc[coord[0], coord[1]] * (len(ccorr_data) - 1))]
        
    # produce the corrected ratiomap by dividing the original
    # ratiomap values by the values in the correction Dataframe
    normalized_ratiomap = (df / correction_df).fillna(0)
    return(normalized_ratiomap)
```

```python
def extend_reference_dfs(num_control_df, denom_control_df):
    # if time points for constitutive correction data and ratio data
    # don't match, interpolates or extrapolates missing time points
    # from existing data
    
    for df in [num_control_df, denom_control_df]:
        existing_reference_times = [float(x) for x in df.columns]
        missing_reference_times = [x for x in times_in_hours if x not in existing_reference_times]
        if not missing_reference_times:
            continue
        else:
            for x in missing_reference_times:
                # use data from the earliest time point if ratio data
                # are from a time point earlier than that
                if x < min(existing_reference_times):
                    df[f'{x}'] = df.loc[:, f'{min(existing_reference_times)}']
                    
                # use data from the latest time point if ratio data
                # are from a time point later than that
                elif x > max(existing_reference_times):
                    df[f'{x}'] = df.loc[:, f'{max(existing_reference_times)}']
                
                # if time point for ratio data falls between two
                # time points present in constitutive correction
                # data, create a weighted average using the two
                # closest time points
                else:
                    first = bisect.bisect_left(existing_reference_times, x) - 1
                    first_data = df.iloc[:, first].dropna().tolist()
                    second = first + 1
                    second_data = df.iloc[:, second].dropna().tolist()
                    diff = existing_reference_times[second] - existing_reference_times[first]
                    first_diff = x - first
                    first_weight = (diff - first_diff) / diff
                    second_weight = 1 - first_weight
                    longer = max([len(x) for x in [first_data, second_data]])
                    data = (pd.Series(resample_list_to(first_data, longer)) * first_weight, pd.Series(resample_list_to(second_data, longer)) * second_weight)
                    weighted_data = list(data[0] + data[1])
                    df[f'{x}'] = pd.Series(weighted_data)
        df = df.reindex(sorted(df.columns), axis = 1)
    return(num_control_df, denom_control_df)
```

```python
def filter_hidden(some_list):
    # removes any hidden files in the input directory from further
    # analysis
    
    new_list = sorted([x for x in some_list if "._" not in x])
    if len(new_list) < len(some_list):
        print(f'Detected {len(some_list) - len(new_list)} hidden files in the input directory\n'
              + f'Hidden image files have been excluded from analysis.')
    return(new_list)
```

```python
def flip_df(df):
    # flips a DataFrame vertically. used when generating "surface-in"
    # profiles
    
    temp_df = df.copy()
    for nans, col in zip(df.isna().sum().tolist(), df.columns):
        temp_df[col] = df[col].shift(nans)
    flipped_df = temp_df.iloc[::-1].reset_index(drop = True)
    flipped_df.index = df.index
    return(flipped_df)
```

```python
def get_control_file(some_fluor):
    # searches directory for spreadsheet that contains constitutive
    # correction data suitable to use for current fluorophore. must
    # match cell line and fluorophore, also attempts to match
    # condition. allows for correction data to be pulled from
    # multiple spreadsheets
    
    try:
        control_files = [f'{path.stem}.xlsx' for path in Path(dir_input).rglob("*.xlsx") if path.stem.startswith("Constitutive data")]
        if len(control_files) > 1:
            sheetnames_lists = [(x, pd.ExcelFile(Path(dir_input, x)).sheet_names) for x in control_files]
            control_files = [x[0] for x in sheetnames_lists if [y for y in x[1] if (y.startswith(celltype.split("-")[0]) and y.endswith(f'{condition}_{some_fluor}'))]]
            if not control_files:
                control_files = [x for x in sheetnames_lists if [y for y in x[1] if (y.startswith(celltype.split("-")[0]) and y.endswith(some_fluor))]]
        control_files = sorted(control_files, key = lambda v: v.upper())
        control_file = pd.ExcelFile(Path(dir_input, control_files[0]))
        print(f'{celltype} {some_fluor} control file: {control_files[0]}')
        return(control_file)
    except:
        print(f'Unable to locate a constitutive data file with a sheet name suitable for {celltype} {condition} {some_fluor}. Constitutive correction disabled.')
        raise ValueError()
```

```python
def get_intensities_intensity(well):
    # "hub" function for getting intensity profiles
    
    # read in image as DataFrames. float64 avoids over/underflow
    # issues
    df = pd.DataFrame(skimage.io.imread(well, as_gray = True)).astype("float64")
    
    # create masked image (foreground), identify centroid, and get
    # foreground coordinates
    df_masked, centroid, foreground_coords = process_image_intensity(df)
    
    if profile_or_comprehensive == "profile":
        
        # get intensity profiles radiating outward from centroid
        intensity_profiles = get_profiles(df_masked, centroid)
        
        # tidy up resulting DataFrame
        intensity_profiles = cleanup_df(intensity_profiles)
        if intensity_profiles.empty:
            return(None)
            
        # determine average profile
        intensity_profile = resample_df(intensity_profiles).mean(axis = 1)
        
    elif profile_or_comprehensive == "comprehensive":
        
        # create a DataFrame to be populated with distance values
        dist_df = pd.DataFrame(index = range(df_masked.shape[0]), columns = range(df_masked.shape[1]))
        
        # determine Euclidean distance between each foreground pixel and
        # centroid. populate distance DataFrame with these values as
        # integers to allow for grouping in next step
        for coord in foreground_coords:
            dist_df.iloc[coord[0], coord[1]] = int(math.dist([coord[0], coord[1]], centroid))
            
        # get the approximate maximum distance. exludes small numbers of
        # pixels that may be separate from the main specimen (e.g., loose
        # cells)
        dmax = np.percentile(dist_df.values[dist_df.notna()], 99.9)
        
        # extract a list of values corresponding to each distance from
        # the centroid of the foreground in 1 pixel increments. build
        # a list of lists, then create a DataFrame from that
        values_by_distance = []
        for i in range(int(dmax + 1)):
            values_by_distance.append(df_masked.values[dist_df.values == i])
        comprehensive_df = pd.DataFrame(values_by_distance)
        if comprehensive_df.empty:
            return(None)
            
        # calculate the mean intensity profile by taking the average of
        # the intensity values at each distance
        intensity_profile = comprehensive_df.mean(axis = 1)
        
    return(intensity_profile)
```

```python
def get_intensities_ratio(num_well, denom_well):
    # "hub" function for getting intensity profiles
    
    # read in paired images as DataFrames. float64 avoids over/
    # underflow issues
    df_num = pd.DataFrame(skimage.io.imread(num_well, as_gray = True)).astype("float64")
    df_denom = pd.DataFrame(skimage.io.imread(denom_well, as_gray = True)).astype("float64")
    
    # create masked images (foreground), identify centroids, and get
    # foreground coordinates
    df_num_masked, df_denom_masked, centroid, foreground_coords = process_image_ratio(df_num, df_denom)
    
    if profile_or_comprehensive == "profile":
        
        # get intensity profiles radiating outward from centroid for
        # each image (numerator and denominator)
        num_profiles = get_profiles(df_num_masked, centroid)
        denom_profiles = get_profiles(df_denom_masked, centroid)
        
        # tidy up resulting DataFrames
        num_profiles, denom_profiles = cleanup_df(num_profiles, denom_profiles)
        if num_profiles.empty or denom_profiles.empty:
            return(None, None)
        
        # determine average profile for each channel
        num_profile = resample_df(num_profiles).mean(axis = 1)
        denom_profile = resample_df(denom_profiles).mean(axis = 1)
        
    elif profile_or_comprehensive == "comprehensive":
        # create DataFrames to be populated with distance values
        
        dist_df = pd.DataFrame(index = range(df_num_masked.shape[0]), columns = range(df_num_masked.shape[1]))
        
        # determine Euclidean distance between each foreground pixel
        # and the respective centroid. populate distance DataFrame
        # with these values as integers to allow for grouping in next
        # step
        for coord in foreground_coords:
            dist_df.iloc[coord[0], coord[1]] = int(math.dist([coord[0], coord[1]], centroid))

        dmax = np.percentile(dist_df.values[dist_df.notna()], 99.9)
        
        # extract a list of values corresponding to each distance from
        # the centroid of the foreground in 1 pixel increments. build
        # lists of lists, then create DataFrames from them
        values_by_distance_num = []
        for i in range(int(dmax + 1)):
            values_by_distance_num.append(df_num_masked.values[dist_df.values == i])
        comprehensive_df_num = pd.DataFrame(values_by_distance_num)
        
        values_by_distance_denom = []
        for i in range(int(dmax + 1)):
            values_by_distance_denom.append(df_denom_masked.values[dist_df.values == i])
        comprehensive_df_denom = pd.DataFrame(values_by_distance_denom)
        
        if comprehensive_df_num.empty or comprehensive_df_denom.empty:
            return(None, None)
            
        # calculate mean intensity profiles by taking the average of
        # the intensity values at each distance
        num_profile = comprehensive_df_num.mean(axis = 1)
        denom_profile = comprehensive_df_denom.mean(axis = 1)
        
    return(num_profile, denom_profile)
```

```python
def get_profiles(df, centroid):
    # algorithm for getting eight intensity profiles radiating from
    # centroid of foreground. see paper (Fig. 1A)
    
    # cardinal and ordinal compass directions
    directions = ["north", "east", "south", "west",
                  "northwest", "northeast", "southwest", "southeast"]
    
    # set up data structure to dump values into
    data = [[direction, [], []] for direction in directions]
    for item in data:
        
        # start with centroid value
        idx = centroid[0]
        col = centroid[1]
        direction = item[0]
        
        # walk from centroid outward until encountering a boundary of
        # the image and record values along the way
        while 0 <= idx < len(df.index) and 0 <= col < len(df.columns):
            item[1].append((idx, col))
            if "north" in direction:
                idx -= 1
            if "south" in direction:
                idx += 1
            if "west" in direction:
                col -= 1
            if "east" in direction:
                col += 1
                
        # get rid of trailing zeroes (background pixels)
        item[2] = np.trim_zeros([df.iloc[coord] for coord in item[1]], trim = "b")
        
    # ordinal directions cover root(2) distance per iteration, versus
    # single pixel per iteration for cardinal directions. thus, need
    # to be resampled to increase number of values by a factor of
    # root(2) without changing shape of curve
    for item in data[4:8]:
        item[2] = resample_list_to(item[2], int(len(item[2]) * math.sqrt(2)))
    
    # aggregate data. if any direction(s) produced an empty list but
    # the other directions had values, this is counted as a list of
    # zeroes
    profiles = [x[2] for x in data]
    valid_profiles = [x for x in profiles if len(x) > 0]
    if len(valid_profiles) < len(profiles):
        avg_profile_length = np.mean([len(x) for x in valid_profiles], dtype = "int")
        zero_list = [0] * avg_profile_length
        while len(valid_profiles) < len(profiles):
            valid_profiles.append([zero_list])
    return(pd.DataFrame(valid_profiles).T)
```

```python
def get_ratiomap(num_well, denom_well):
    # creates ratio maps from input images. see paper (Fig. 1G)
    
    # read input images as DataFrames
    df_num = pd.DataFrame(skimage.io.imread(num_well, as_gray = True))
    df_denom = pd.DataFrame(skimage.io.imread(denom_well, as_gray = True))
    
    # minor blur and background subtraction
    df_num_gaussian = pd.DataFrame(gaussian_filter(df_num, sigma = 1))
    bg_num = np.percentile(df_num_gaussian, bg_percentile)
    df_num_gaussian_bgsub = (df_num_gaussian - bg_num).clip(lower = 0)
    df_denom_gaussian = pd.DataFrame(gaussian_filter(df_denom, sigma = 1))
    bg_denom = np.percentile(df_denom_gaussian, bg_percentile)
    df_denom_gaussian_bgsub = (df_denom_gaussian - bg_denom).clip(lower = 0)

    # create masks for the foreground of each channel, accounts for small numbers
    # of extremely bright pixels
    sat_num = np.percentile(df_num_gaussian_bgsub, 98)
    df_num_gaussian_bgsub_robust_mask = df_num_gaussian_bgsub.clip(upper = sat_num)
    thresh_num = skimage.filters.threshold_otsu(df_num_gaussian_bgsub_robust_mask.values)
    sat_denom = np.percentile(df_denom_gaussian_bgsub, 98)
    df_denom_gaussian_bgsub_robust_mask = df_denom_gaussian_bgsub.clip(upper = sat_denom)
    thresh_denom = skimage.filters.threshold_otsu(df_denom_gaussian_bgsub_robust_mask.values)
    
    # calculate SNR as threshold / background
    SNR_num = thresh_num / bg_num
    SNR_denom = thresh_denom / bg_denom
    
    # compare SNRs to determine which channel is better to use for
    # masking to identify the foreground. numerator floor set to 0
    # and denominator floor set to denominator threshold
    if SNR_num > SNR_denom:
        df_num_masked = df_num_gaussian_bgsub.where(df_num_gaussian_bgsub_robust_mask > thresh_num, other = 0)
        df_denom_masked = df_denom_gaussian_bgsub.where(df_num_gaussian_bgsub_robust_mask > thresh_num, other = thresh_denom)
    elif SNR_num < SNR_denom:
        df_num_masked = df_num_gaussian_bgsub.where(df_denom_gaussian_bgsub_robust_mask > thresh_denom, other = 0)
        df_denom_masked = df_denom_gaussian_bgsub.where(df_denom_gaussian_bgsub_robust_mask > thresh_denom, other = thresh_denom)
    elif SNR_num == SNR_denom:
        # if SNRs are equal, uses a combined mask
        df_num_masked = df_num_gaussian_bgsub.where((df_num_gaussian_bgsub_robust_mask > thresh_num) | (df_denom_gaussian_bgsub_robust_mask > thresh_denom), other = 0)
        df_denom_masked = df_denom_gaussian_bgsub.where((df_num_gaussian_bgsub_robust_mask > thresh_num) | (df_denom_gaussian_bgsub_robust_mask > thresh_denom), other = thresh_denom)
    
    # image math + minor blur
    df_ratiomap = df_num_masked / df_denom_masked
    df_ratiomap = pd.DataFrame(gaussian_filter(df_ratiomap, sigma = 1))
    
    return(df_ratiomap)
```

```python
def get_reference_dfs(num_fluor, denom_fluor):
    # pulls control data from control files
    
    # control files may contain any number of combinations of cell
    # type, fluorophore, condition, etc. this function pulls the most
    # appropriate sheet for the data being analyzed. must match cell
    # line and fluorophore, attempts to match condition as well. if
    # data is available for a different cell line but the same
    # fluorophore and same condition, that data will be substituted.
    # user is notified of which files and sheets are used for what,
    # so anything that doesn't match with what the user wants can be
    # identified easily
    num_celltype = celltype.replace(biosensor, num_fluor)
    denom_celltype = celltype.replace(biosensor, denom_fluor)
    try:
        try:
            num_sheet = f'{num_celltype}_{condition}_{num_fluor}'[:31]
            control_file_num = get_control_file(num_fluor)
            num_control_df = pd.read_excel(control_file_num, num_sheet)
        except:
            num_fluor_sheets = [str(x) for x in control_file_num.sheet_names if str(x).startswith(celltype.split("-")[0]) and str(x).endswith(num_fluor)]
            num_fluor_condition_sheets = [x for x in num_fluor_sheets if f'_{condition}_' in x]
            num_fluor_conditions_nonself_sheets = [x for x in num_fluor_condition_sheets if not x.startswith(celltype)]
            if len(num_fluor_sheets) >= 1:
                num_sheet = num_fluor_sheets[0][:31]
                if len(num_fluor_condition_sheets) >= 1:
                    num_sheet = num_fluor_condition_sheets[0][:31]
                    if len (num_fluor_conditions_nonself_sheets) >= 1:
                        num_sheet = num_fluor_conditions_nonself_sheets[0][:31]
        try:
            denom_sheet = f'{denom_celltype}_{condition}_{denom_fluor}'[:31]
            control_file_denom = get_control_file(denom_fluor)
            denom_control_df = pd.read_excel(control_file_denom, denom_sheet)
        except:
            denom_fluor_sheets = [str(x) for x in control_file_denom.sheet_names if str(x).startswith(celltype.split("-")[0]) and str(x).endswith(denom_fluor)]
            denom_fluor_condition_sheets = [x for x in denom_fluor_sheets if f'_{condition}_' in x]
            denom_fluor_conditions_nonself_sheets = [x for x in denom_fluor_condition_sheets if not x.startswith(celltype)]
            if len(denom_fluor_sheets) >= 1:
                denom_sheet = denom_fluor_sheets[0][:31]
                if len(denom_fluor_condition_sheets) >= 1:
                    denom_sheet = denom_fluor_condition_sheets[0][:31]
                    if len (denom_fluor_conditions_nonself_sheets) >= 1:
                        denom_sheet = denom_fluor_conditions_nonself_sheets[0][:31]
        
        num_control_df = pd.read_excel(control_file_num, num_sheet)
        denom_control_df = pd.read_excel(control_file_denom, denom_sheet)
        num_control_df.columns = [col.split(" = ")[1] for col in num_control_df.columns]
        denom_control_df.columns = [col.split(" = ")[1] for col in denom_control_df.columns]
        
        # interpolates/extrapolates in case data being analyzed have
        # time points that are not present in control data
        extend_reference_dfs(num_control_df, denom_control_df)
        print('')
        print(f'Normalizing ({celltype} {condition} {num_fluor}) data using constitutive data sheet ({num_sheet}).')
        print(f'Normalizing ({celltype} {condition} {denom_fluor}) data using constitutive data sheet ({denom_sheet}).')
        print('')
        return(num_control_df, denom_control_df)
        
    # if pulling control data fails, user will be notified and
    # analysis will run without constitutive correction
    except:
        print(f'Proceeding without correction for {celltype}...')
        global cc_toggle
        cc_toggle = False
        print('')
        return(None, None)
```

```python
def get_regional_data(df):
    # bins and quantifies regional data based on number of regions
    # assigned by user (quantification_regions). see paper (Fig. 1F)
    # input DataFrames are set up as one column (profile) per
    # replicate
    
    # handles instances where number of desired regions exceeds
    # number of values in profile. if this is the case, the regional
    # data looks identical to the unbinned data when output
    if len(df.index) < quantification_regions:
        region_size = 1
        new_regions = [f'region {i + 1}' for i in range(len(df.index))]
    
    # otherwise, sets the appropriate bin size
    else:
        region_size = len(df.index) / quantification_regions
        
    # separates replicates (regional data is calculated per-
    # replicate)
    value_lists = [df[x].dropna().tolist() for x in df.columns]
    regional_df = pd.DataFrame()
    for c, value_list in enumerate(value_lists):
        regional_value_list = []
        
        # segments full profile based on bin size and takes mean
        for i in range(quantification_regions):
            region_values = value_list[int(region_size * i):int(region_size * (i + 1))]
            region_mean = sum(region_values) / len(region_values)
            regional_value_list.append(region_mean)
            i += 1
        
        # extracts values into new DataFrame
        regional_df[df.columns[c]] = regional_value_list
    
    # tidy up if necessary
    regional_df = cleanup_df(regional_df)
    if len(df.index) < quantification_regions:
        regional_df.index = new_regions
    else:
        regional_df.index = regions
    return(regional_df)
```

```python
def get_time_reference(num_control_df, denom_control_df, time_hours):
    # gets constitutive correction data for a given time point from
    # the control DataFrames (pulled from .xlxs by get_reference_dfs)
    
    ref_data_num = num_control_df[f'{time_hours}'].dropna().tolist()
    ref_data_denom = denom_control_df[f'{time_hours}'].dropna().tolist()
    avg_reference_length = int((len(ref_data_num) + len(ref_data_denom)) / 2)
    resampled_num = pd.Series(resample_list_to(ref_data_num, avg_reference_length))
    resampled_denom = pd.Series(resample_list_to(ref_data_denom, avg_reference_length))
    reference_data = resampled_num / resampled_denom
    reference_data = reference_data / reference_data.min()
    return(reference_data)
```

```python
def get_well_list(replicate_flist):
    # creates a list of paired files for image processing later on
    
    wells = []
    for terminal_file in replicate_flist:
        f_name_num = f'{(Path(dir_input, "_".join(terminal_file)))}.tif'
        f_name_denom = f_name_num.replace(f'_{num_fluor}_', f'_{denom_fluor}_')
        wells.append((f_name_num, f_name_denom))
    return(wells)
```

```python
def make_graphs_2d_intensity(dfs, dir_2d, direction, data_format, mirrored = False):
    # creates 2D graphs. see paper for details (Fig. 1, 2)
    
    # lots can be changed in here to suit user preference. default
    # setup is fairly robust and will generally present data nicely.
    # users may wish to assign specific values for ymax and xmax (and
    # xmin, if mirrored) to ensure graphs comparing conditions or
    # cell lines are plotted on comparable axes.
    
    # honestly, this function is a real mess because graphing is hard
    # and it works as-is. going back to tidy it up hasn't been a
    # priority. same applies for 3D graph function. sorry!
    
    if graphs_2d:
        try_dir(dir_2d)
        df = dfs[0].copy()
        df_error = dfs[1].copy()
        
        # set depending on optional data formats
        if direction == "centre-out":
            xaxis_label = f'Distance from core ({distance_unit})'
        elif direction == "surface-in":
            xaxis_label = f'Distance from surface ({distance_unit})'
            
        if data_format[1] == "raw":
            yaxis_label = f'{fluor} intensity (RFU)'
        elif data_format[1] == "relative":
            yaxis_label = f'Rel. {fluor} intensity'
        elif data_format[1] == "percentage":
            yaxis_label = f'Rel. {fluor} intensity\n(% of max)'

        # sets axis limits based on data, but can be set manually
        # for reasons noted above
        ymax = (df + df_error).max().max() * 1.05
        df['microns'] = df.index * dist_per_px
        if mirrored:
            df['microns'] = df['microns'] - (df['microns'].max() / 2)
        df_error['microns'] = df['microns'].copy()
        
        xmin = math.floor(df['microns'].min() / 100) * 100
        xmax = math.ceil(df['microns'].max() / 100) * 100
        
        df = df.melt(id_vars = ['microns'])
        df_error = df_error.melt(id_vars = ['microns'])
        df['variable'] = df['variable'].str.replace("_", " ")
        df_error['variable'] = df['variable'].str.replace("_", " ")
        temp = df['variable'].str.split(" ", n = 2, expand = True)
        df['time only'] = temp[2]
        
        dir_2d_time = Path(dir_2d, "2D by time point (with error)", data_format[0])
        try_dir(dir_2d_time)
        
        # uses middle 70% of graphs_cmap
        colors = plt.colormaps[graphs_cmap](np.linspace(0.15, 0.85, len(set(df['variable']))))
        sns.set_theme(style = "ticks", font_scale = 4.5)
        
        # creates a single plot for each time point, includes error
        # values as well. see paper (Fig. 1B)
        fig = plt.figure(figsize = (12, 9))
        for color, time in enumerate(df['variable'].unique()):
            data = df.loc[df['variable'] == time]
            error = df_error.loc[df_error['variable'] == time]
            time_only = time.split(" ")[2]
            
            ax1 = fig.add_subplot(position = [0.3, 0.3, 0.65, 0.65])
            ax1 = sns.lineplot(data = data,
                               x = 'microns',
                               y = 'value', 
                               color = colors[color], 
                               linewidth = 2, 
                               zorder = 5, 
                               legend = False)

            data['lower'] = data['value'] - error['value']
            data['upper'] = data['value'] + error['value']

            if not data['lower'].equals(data['upper']):
                ax1 = sns.lineplot(data = data, x = 'microns', y = 'lower', color = colors[color], linewidth = 1, zorder = 3, legend = False)
                ax1 = sns.lineplot(data = data, x = 'microns', y = 'upper', color = colors[color], linewidth = 1, zorder = 3, legend = False)
                ax1.fill_between(data['microns'], data['lower'], data['upper'], color = colors[color], alpha = 0.4)

            ax1.set(xlim = (xmin, xmax), ylim = (0, ymax))
            ax1.spines['left'].set_linewidth(2)
            ax1.spines['bottom'].set_linewidth(2)
            ax1.tick_params(direction = 'out', labelsize = 32, length = 6, width = 2, bottom = True, left = True, right = False, top = False)
            xticks = np.linspace(xmin, xmax, 5)
            if mirrored:
                ax1.set_xticks(xticks, labels = [str(abs(int(tick))) for tick in xticks], fontweight = 'demibold')
            else:
                ax1.set_xticks(xticks, labels = [str(int(tick)) for tick in xticks], fontweight = 'demibold')
            ax1.set_yticks(ax1.get_yticks(), labels = ax1.get_yticklabels(), fontweight = 'demibold')
            ax1.set_xlabel(xaxis_label, labelpad = 12, fontsize = 34, fontweight = 'demibold')
            ax1.set_ylabel(yaxis_label, labelpad = 12, fontsize = 34, fontweight = 'demibold')
            sns.despine(offset = 10)
            if mirrored:
                plt.savefig(Path(dir_2d_time, f'2D {celltype} {condition} {fluor} {data_format[0]} - mirrored intensity profile for {time_only} (with error).png'), dpi = 300, bbox_inches = "tight")
            else:
                plt.savefig(Path(dir_2d_time, f'2D {celltype} {condition} {fluor} {data_format[0]} - {direction} intensity profile for {time_only} (with error).png'), dpi = 300, bbox_inches = "tight")
            plt.clf()
        plt.close()
        
        # creates a single plot with all time points stacked. see
        # paper (Fig. 1C)
        stacked_colors = np.flip(colors, axis = 0)
        fig = plt.figure(figsize = (12, 9))
        ax1 = fig.add_subplot(position = [0.3, 0.3, 0.65, 0.65])
        ax1 = sns.lineplot(data = df.sort_values(['time only'], ascending = False),
                           x = 'microns',
                           y = 'value',
                           hue = 'time only',
                           estimator = None,
                           palette = stacked_colors,
                           linewidth = 2,
                           legend = "full")

        ax1.set(xlim = (xmin, xmax), ylim = (0, ymax))
        ax1.spines['left'].set_linewidth(2)
        ax1.spines['bottom'].set_linewidth(2)
        ax1.tick_params(direction = 'out', labelsize = 32, length = 6, width = 2, bottom = True, left = True, right = False, top = False)
        xticks = np.linspace(xmin, xmax, 5)
        if mirrored:
            ax1.set_xticks(xticks, labels = [str(abs(int(tick))) for tick in xticks], fontweight = 'medium')
        else:
            ax1.set_xticks(xticks, labels = [str(int(tick)) for tick in xticks], fontweight = 'medium')
        ax1.set_yticks(ax1.get_yticks(), labels = ax1.get_yticklabels(), fontweight = 'medium')
        ax1.set_xlabel(xaxis_label, labelpad = 12, fontsize = 34, fontweight = 'demibold')
        ax1.set_ylabel(yaxis_label, labelpad = 12, fontsize = 34, fontweight = 'demibold')
        sns.despine(offset = 10)
        handles, labels = ax1.get_legend_handles_labels()
        ax1.legend()
        ax1.get_legend().set_visible(False)
        if mirrored:
            plt.savefig(Path(dir_2d, f'2D {celltype} {condition} {fluor} {data_format[0]} - stacked mirrored intensity profiles.png'), dpi = 300, bbox_inches = "tight")
        else:
            plt.savefig(Path(dir_2d, f'2D {celltype} {condition} {fluor} {data_format[0]} - stacked {direction} intensity profiles.png'), dpi = 300, bbox_inches = "tight")
        legend = plt.figure()
        legend.legend(handles[::-1],
                      labels[::-1],
                      ncols = math.ceil(len(labels) / 24),
                      fontsize = 9,
                      title = (f'{celltype}\n{condition} {fluor}'),
                      title_fontsize = 9,
                      labelspacing = 0.2,
                      facecolor = 'white',
                      framealpha = 0.4)
        legend.savefig(Path(dir_2d, f'{celltype} {condition} {fluor} legend.png'), dpi = 300, bbox_inches = "tight")
        plt.close()
```

```python
def make_graphs_3d_intensity(dfs, dir_3d, direction, data_format, mirrored = False):
    # creates 3D graphs. see paper for details (Fig. 1)
    
    # lots can be changed in here to suit user preference. default
    # setup is fairly robust and will generally present data nicely.
    # users may wish to assign specific values for ymax and xmax (and
    # xmin, if mirrored) to ensure graphs comparing conditions or
    # cell lines are plotted on comparable axes.
    
    # honestly, this function is a real mess because graphing is hard
    # and it works as-is. going back to tidy it up hasn't been a
    # priority. same applies for 2D graph function. sorry!
    
    if graphs_3d:
        try_dir(dir_3d)
        df = dfs[0].copy()
        
        # set depending on optional data formats
        if direction == "centre-out":
            xaxis_label = f'Distance from core ({distance_unit})'
        elif direction == "surface-in":
            xaxis_label = f'Distance from surface ({distance_unit})'
            
        if data_format[1] == "raw":
            zaxis_label = f'{fluor} intensity (RFU)'
        elif data_format[1] == "relative":
            zaxis_label = f'Rel. {fluor} intensity'
        elif data_format[1] == "percentage":
            zaxis_label = f' Rel.{fluor} intensity\n(% of max)'
            
        # determines labels for time points
        df.columns = [x.split("_")[-1] for x in df.columns]
        times_in_hours = [(24 * int(time.split("_")[-1].split("d")[0]) + int(time.split("d")[1].split("h")[0]) + (int(time.split("h")[1].split("m")[0]) / 60)) for time in df.columns]
        
        # sets axis limits based on data, but can be set manually
        # for reasons noted above
        zmax = df.max().max() * 1.1
        
        if mirrored:
            df['microns'] = (df.index - (len(df.index) / 2)) * dist_per_px
        else:
            df['microns'] = df.index * dist_per_px
        
        xmin = math.floor(df['microns'].min() / 100) * 100
        xmax = math.ceil(df['microns'].max() / 100) * 100
        xvals = np.linspace(df['microns'].min(), df['microns'].max(), len(df.index))
        df = df.melt(id_vars = ['microns'])
        df = df.fillna(0)

        times_to_plot = list(df['variable'].unique())
        verts = [polygon_under_graph(xvals, df.loc[df['variable'] == time]['value']) for time in times_to_plot]
        
        # uses middle 70% of graphs_cmap
        facecolors = plt.colormaps[graphs_cmap](np.linspace(0.15, 0.85, len(verts)))
        poly = PolyCollection(verts, facecolors = facecolors, alpha = 0.7, edgecolors = ('black'))
        
        # creates a single 3D graph that displays all time points.
        # see paper (Fig. 1E)
        fig = plt.figure(figsize = (12, 9))
        ax3d = fig.add_subplot(projection = '3d')
        fig.subplots_adjust(left = 0, right = 1, bottom = 0, top = 1)
        ax3d.set_box_aspect(aspect = (5,5,4))
        ax3d.add_collection3d(poly, zs = times_in_hours, zdir = 'y')
        # ax3d.set(xlim = (xmin * 1.05, xmax * 1.05), ylim = (times_in_hours[0] / 3, (times_in_hours[-1] + times_in_hours[0])), zlim = (0, zmax))
        ax3d.set(xlim = (xmin * 1.05, xmax * 1.05), ylim = (0, (times_in_hours[-1] + times_in_hours[1])), zlim = (0, zmax)) ######################
        
        ax3d.tick_params(direction = 'out', labelsize = 22)
        ax3d.tick_params(axis = "z", pad = -3)
        ax3d.tick_params(axis = "x", pad = 10)
        ax3d.tick_params(axis = "y", pad = 10)
        xticks = np.linspace(xmin, xmax, 5)
        if mirrored:
            ax3d.set_xticks(xticks, labels = [str(abs(int(tick))) for tick in xticks], va = 'baseline', fontweight = 'medium')
        else:
            ax3d.set_xticks(xticks, labels = [str(int(tick)) for tick in xticks], va = 'baseline', fontweight = 'medium')
        yticks = [times_in_hours[y] for y in [int(x) for x in np.linspace(0, len(times_in_hours) - 1, 6)]]
        ax3d.set_yticks(yticks, labels = [str(int(tick)) for tick in yticks], va = 'baseline', fontweight = 'medium')
        ax3d.set_zticks(ax3d.get_zticks(), labels = ax3d.get_zticklabels(), va = 'center', ha = 'right', fontweight = 'medium')

        ax3d.view_init(elev = 10, azim = -45)
        ax3d.zaxis._axinfo['juggled'] = (1,2,0)
        
        ax3d.zaxis.set_rotate_label(False)
        ax3d.set_xlabel(xaxis_label, fontsize = 26, labelpad = 12, fontweight = 'demibold')
        ax3d.set_ylabel("Elapsed time (h)", fontsize = 26, labelpad = 12, fontweight = 'demibold')
        zlabelpad = max([len(x.get_text()) for x in ax3d.get_zticklabels()]) * 5
        ax3d.set_zlabel(zaxis_label, fontsize = 26, labelpad = zlabelpad, rotation = 90, fontweight = 'demibold')
        
        if mirrored:
            plt.savefig(Path(dir_3d, f'3D {celltype} {condition} {fluor} {data_format[0]} - mirrored 3D (front).png'), dpi = 300, edgecolor = 'black')
        else:    
            plt.savefig(Path(dir_3d, f'3D {celltype} {condition} {fluor} {data_format[0]} - {direction} 3D (front outer).png'), dpi = 300, edgecolor = 'black')
            ax3d.view_init(elev = 10, azim = -135)
            ax3d.zaxis._axinfo['juggled'] = (0,2,1)
            plt.savefig(Path(dir_3d, f'3D {celltype} {condition} {fluor} {data_format[0]} - {direction} 3D (front inner).png'), dpi = 300, edgecolor = 'black')
        
        ax3d.view_init(elev = 10, azim = 45)
        ax3d.zaxis._axinfo['juggled'] = (0,2,0)
        if mirrored:
            plt.savefig(Path(dir_3d, f'3D {celltype} {condition} {fluor} {data_format[0]} - mirrored 3D (rear).png'), dpi = 300, edgecolor = 'black')
        else:
            plt.savefig(Path(dir_3d, f'3D {celltype} {condition} {fluor} {data_format[0]} - {direction} 3D (rear outer).png'), dpi = 300, edgecolor = 'black')
            ax3d.view_init(elev = 10, azim = 135)
            ax3d.zaxis._axinfo['juggled'] = (1,2,1)
            plt.savefig(Path(dir_3d, f'3D {celltype} {condition} {fluor} {data_format[0]} - {direction} 3D (rear inner).png'), dpi = 300, edgecolor = 'black')
        plt.close()
```

```python
def make_graphs_3d_ratio(dfs, dir_3d, direction, data_format, ccorr_status, mirrored = False):
    # creates 3D graphs. see paper for details (Fig. 1)
    
    # lots can be changed in here to suit user preference. default
    # setup is fairly robust and will generally present data nicely.
    # users may wish to assign specific values for ymax and xmax (and
    # xmin, if mirrored) to ensure graphs comparing conditions or
    # cell lines are plotted on comparable axes.
    
    # honestly, this function is a real mess because graphing is hard
    # and it works as-is. going back to tidy it up hasn't been a
    # priority. same applies for 2D graph function. sorry!
    
    if graphs_3d:
        try_dir(dir_3d)
        df = dfs[0].copy()
        
        # set depending on optional data formats
        if direction == "centre-out":
            xaxis_label = f'Distance from core ({distance_unit})'
        elif direction == "surface-in":
            xaxis_label = f'Distance from surface ({distance_unit})'
            
        if data_format[1] == "raw":
            zaxis_label = f'{ratio_expression} ratio'
        elif data_format[1] == "relative":
            zaxis_label = f'Rel. {ratio_expression} ratio'
        elif data_format[1] == "percentage":
            zaxis_label = f'Rel. {ratio_expression} ratio\n(% of max)'
            
        # determines labels for time points
        df.columns = [x.split("_")[-1] for x in df.columns]
        times_in_hours = [(24 * int(time.split("_")[-1].split("d")[0]) + int(time.split("d")[1].split("h")[0]) + (int(time.split("h")[1].split("m")[0]) / 60)) for time in df.columns]
        
        # sets axis limits based on data, but can be set manually
        # for reasons noted above
        zmax = df.max().max() * 1.1
        
        if mirrored:
            df['microns'] = (df.index - (len(df.index) / 2)) * dist_per_px
        else:
            df['microns'] = df.index * dist_per_px
            
        xmin = math.floor(df['microns'].min() / 100) * 100
        xmax = math.ceil(df['microns'].max() / 100) * 100
        xvals = np.linspace(df['microns'].min(), df['microns'].max(), len(df.index))
        df = df.melt(id_vars = ['microns'])
        df = df.fillna(0)
        
        times_to_plot = list(df['variable'].unique())
        verts = [polygon_under_graph(xvals, df.loc[df['variable'] == time]['value']) for time in times_to_plot]
        
        # uses middle 70% of graphs_cmap
        facecolors = plt.colormaps[graphs_cmap](np.linspace(0.15, 0.85, len(verts)))
        poly = PolyCollection(verts, facecolors = facecolors, alpha = 0.7, edgecolors = ('black'))
        
        # creates a single 3D graph that displays all time points.
        # see paper (Fig. 1E)
        fig = plt.figure(figsize = (12, 9))
        ax3d = fig.add_subplot(projection = '3d')
        fig.subplots_adjust(left = 0, right = 1, bottom = 0, top = 1)
        ax3d.set_box_aspect(aspect = (5,5,4))
        ax3d.add_collection3d(poly, zs = times_in_hours, zdir = 'y')
        # ax3d.set(xlim = (xmin * 1.05, xmax * 1.05), ylim = (times_in_hours[0] / 3, (times_in_hours[-1] + times_in_hours[0])), zlim = (0, zmax))
        ax3d.set(xlim = (xmin * 1.05, xmax * 1.05), ylim = (0, (times_in_hours[-1] + times_in_hours[1])), zlim = (0, zmax)) ######################
        
        ax3d.tick_params(direction = 'out', labelsize = 22)
        ax3d.tick_params(axis = "z", pad = -3)
        ax3d.tick_params(axis = "x", pad = 10)
        ax3d.tick_params(axis = "y", pad = 10)
        xticks = np.linspace(xmin, xmax, 5)
        if mirrored:
            ax3d.set_xticks(xticks, labels = [str(abs(int(tick))) for tick in xticks], va = 'baseline', fontweight = 'medium')
        else:
            ax3d.set_xticks(xticks, labels = [str(int(tick)) for tick in xticks], va = 'baseline', fontweight = 'medium')
        yticks = sorted(set([times_in_hours[y] for y in [int(x) for x in np.linspace(0, len(times_in_hours) - 1, 6)]]))
        ax3d.set_yticks(yticks, labels = [str(int(tick)) for tick in yticks], va = 'baseline', fontweight = 'medium')
        ax3d.set_zticks(ax3d.get_zticks(), labels = ax3d.get_zticklabels(), va = 'center', ha = 'right', fontweight = 'medium')
        
        ax3d.view_init(elev = 10, azim = -45)
        ax3d.zaxis._axinfo['juggled'] = (1,2,0)
        
        ax3d.zaxis.set_rotate_label(False)
        ax3d.set_xlabel(xaxis_label, fontsize = 26, labelpad = 12, fontweight = 'demibold')
        ax3d.set_ylabel("Elapsed time (h)", fontsize = 26, labelpad = 12, fontweight = 'demibold')
        zlabelpad = max([len(x.get_text()) for x in ax3d.get_zticklabels()]) * 5
        ax3d.set_zlabel(zaxis_label, fontsize = 26, labelpad = zlabelpad, rotation = 90, fontweight = 'demibold')
        
        if mirrored:
            plt.savefig(Path(dir_3d, f'3D {celltype} {condition}, {ccorr_status} {data_format[0]} - mirrored 3D (front).png'), dpi = 300, edgecolor = 'black')
        else:
            plt.savefig(Path(dir_3d, f'3D {celltype} {condition}, {ccorr_status} {data_format[0]} - {direction} 3D (front outer).png'), dpi = 300, edgecolor = 'black')
            ax3d.view_init(elev = 10, azim = -135)
            ax3d.zaxis._axinfo['juggled'] = (0,2,1)
            plt.savefig(Path(dir_3d, f'3D {celltype} {condition}, {ccorr_status} {data_format[0]} - {direction} 3D (front inner).png'), dpi = 300, edgecolor = 'black')
            
        ax3d.view_init(elev = 10, azim = 45)
        ax3d.zaxis._axinfo['juggled'] = (0,2,0)
        if mirrored:
            plt.savefig(Path(dir_3d, f'3D {celltype} {condition}, {ccorr_status} {data_format[0]} - mirrored 3D (rear).png'), dpi = 300, edgecolor = 'black')
        else:
            plt.savefig(Path(dir_3d, f'3D {celltype} {condition}, {ccorr_status} {data_format[0]} - {direction} 3D (rear outer).png'), dpi = 300, edgecolor = 'black')
            ax3d.view_init(elev = 10, azim = 135)
            ax3d.zaxis._axinfo['juggled'] = (1,2,1)
            plt.savefig(Path(dir_3d, f'3D {celltype} {condition}, {ccorr_status} {data_format[0]} - {direction} 3D (rear inner).png'), dpi = 300, edgecolor = 'black')
        plt.close()
```

```python
def make_rmaps_per_time():
    # creates ratio maps with colour scaling based on the data for a
    # given cell line, condition, and time point. essentially
    # isolates spatial information, but not useful for comparing
    # between times, conditions, cell lines, etc.
    
    info = [x for x in rmaps_reference_info if x[0] == celltype and x[1] == condition]
    reference_df = info[0][2]
    func_rmaps_cmap = info[0][4]
    func_num_fluor = info[0][5]
    func_denom_fluor = info[0][6]
    func_ratio_expression = f'{func_num_fluor}/{func_denom_fluor}'
    func_cc_toggle = info[0][-1]
    dir_rmaps_per_time = Path(dir_celltype_condition, f'ratio-mapped images ({func_rmaps_cmap.split(".")[-1]})', "scaled for each time point")
    try_dir(dir_rmaps_per_time)
    
    # ratio map colormap is taken but lowest value replaced with
    # black to force contrast with background. for this reason, it is
    # generally best not to use colormaps with black on the low end
    # if low ratios are expected near the surface of the spheroid
    # (e.g., for a hypoxia biosensor)
    colors = plt.colormaps[func_rmaps_cmap](np.linspace(0, 1, 2**16))
    black = np.array([0, 0, 0, 1])
    colors[:1, :] = black
    newcmap = ListedColormap(colors)
    
    # if correction is applied to ratio profiles, it is also
    # applied to ratio maps
    if constitutive_correction and func_cc_toggle:
        reference_df_ccorr = info[0][7]
        df_ccorr_data = info[0][8]
        
    print(f'Ratio-mapping images (scaling per time point), colormap = {func_rmaps_cmap}...')
    
    # if using rmaps_sample, randomly samples the data set and ratio
    # maps the same speciments at each time point (nice for
    # representative images)
    samples_assigned = False
    for time in times:
        time_flist = [item for item in condition_flist if (item[6] == time) and (item[2] == func_num_fluor)]
            
        if rmaps_sample and not samples_assigned:
            if rmaps_sample_wells < len(time_flist):
                time_flist_samples = random.sample(time_flist, rmaps_sample_wells)
                samples_assigned = True
        elif rmaps_sample and samples_assigned:
            for x in time_flist_samples:
                x[6] = time
                
        # set colour scaling
        time_data = reference_df[f'{celltype}_{condition}_{time}']
        norm_time_data = time_data / time_data.min()
        norm_max = norm_time_data.max()
        dir_rmaps_out = Path(dir_rmaps_per_time, "no correction", f'{time}')
        try_dir(dir_rmaps_out)
        if constitutive_correction and func_cc_toggle:
            ccorr_data = df_ccorr_data[f'{celltype}_{condition}_{time}'].dropna().tolist()
            time_data_ccorr = reference_df_ccorr[f'{celltype}_{condition}_{time}']
            norm_time_data_ccorr = time_data_ccorr / time_data_ccorr.min()
            norm_max_ccorr = norm_time_data_ccorr.max()
            dir_rmaps_out_ccorr = Path(dir_rmaps_per_time, "with correction", f'{time}')
            try_dir(dir_rmaps_out_ccorr)
            
        print(f'Ratio-mapping {celltype}, {condition}, {time}')
        
        # use sampled specimens (wells) if applicable, otherwise
        # generate ratio maps for whole data set (can take a very
        # long time)
        if samples_assigned:
            wells = get_well_list(time_flist_samples)
        else:
            wells = get_well_list(time_flist)
            
        # generate ratio map, correct it if constitutive correction
        # is enabled
        for num_well, denom_well in wells:
            try:
                df_ratiomap = get_ratiomap(num_well, denom_well)
                if constitutive_correction and func_cc_toggle:
                    df_ratiomap_ccorr = constitutive_correct_ratiomap(df_ratiomap, ccorr_data)
                    norm_df_ratiomap_ccorr = df_ratiomap_ccorr / time_data_ccorr.min()
                    write_ratiomap(norm_df_ratiomap_ccorr, dir_rmaps_out_ccorr, norm_max_ccorr, newcmap, num_well, "with correction")
                norm_df_ratiomap = df_ratiomap / time_data.min()
                write_ratiomap(norm_df_ratiomap, dir_rmaps_out, norm_max, newcmap, num_well, "no correction")
            except KeyboardInterrupt:
                raise KeyboardInterrupt()
            except:
                # if images for specific time points are missing,
                # user will be notified by the whole thing won't
                # crash
                print(f'Unable to generate a ratio map for {num_well}, {denom_well} - file(s) may be missing.')
                continue
                
        # generates a colorbar that corresponds to the scaling of the
        # colormap
        if constitutive_correction and func_cc_toggle:
            write_cbar(time_data_ccorr.min(), time_data_ccorr.max(), dir_rmaps_out_ccorr, newcmap, func_ratio_expression, f'{celltype} {condition} {time}', "with correction")
        write_cbar(time_data.min(), time_data.max(), dir_rmaps_out, newcmap, func_ratio_expression, f'{celltype} {condition} {time}', "no correction")
```

```python
def make_rmaps_compare_times():
    # creates ratio maps with colour scaling based on the data for a
    # given cell line and condition. colour scales across all time
    # points. useful for comparing between times, but not between
    # conditions, cell lines, etc.
    
    info = [x for x in rmaps_reference_info if x[0] == celltype and x[1] == condition]
    reference_df = info[0][2]
    func_rmaps_cmap = info[0][4]
    func_num_fluor = info[0][5]
    func_denom_fluor = info[0][6]
    func_ratio_expression = f'{func_num_fluor}/{func_denom_fluor}'
    func_cc_toggle = info[0][-1]
    dir_rmaps_celltype_condition = Path(dir_celltype_condition, f'ratio-mapped images ({func_rmaps_cmap.split(".")[-1]})', "scaled for all time points")
    try_dir(dir_rmaps_celltype_condition)
    
    # ratio map colormap is taken but lowest value replaced with
    # black to force contrast with background. for this reason, it is
    # generally best not to use colormaps with black on the low end
    # if low ratios are expected near the surface of the spheroid
    # (e.g., for a hypoxia biosensor)
    colors = plt.colormaps[func_rmaps_cmap](np.linspace(0, 1, 2**16))
    black = np.array([0, 0, 0, 1])
    colors[:1, :] = black
    newcmap = ListedColormap(colors)
    
    # set colour scaling
    norm_df = reference_df / reference_df.min().min()
    norm_max = norm_df.max().max()
    
    # if correction is applied to ratio profiles, it is also
    # applied to ratio maps
    if constitutive_correction and func_cc_toggle:
        reference_df_ccorr = info[0][7]
        df_ccorr_data = info[0][8]
        norm_df_ccorr = reference_df_ccorr / reference_df_ccorr.min().min()
        norm_max_ccorr = norm_df_ccorr.max().max()
        
    print(f'Ratio-mapping images (scaling for {celltype} {condition}), colormap = {func_rmaps_cmap}...')
    
    # if using rmaps_sample, randomly samples the data set and ratio
    # maps the same speciments at each time point (nice for
    # representative images)
    samples_assigned = False
    for time in times:
        time_flist = [item for item in condition_flist if (item[6] == time) and (item[2] == func_num_fluor)]
        if not time_flist:
            continue
            
        if rmaps_sample and not samples_assigned:
            if rmaps_sample_wells < len(time_flist):
                time_flist_samples = random.sample(time_flist, rmaps_sample_wells)
                samples_assigned = True
        elif rmaps_sample and samples_assigned:
            for x in time_flist_samples:
                x[6] = time
                
        dir_rmaps_out = Path(dir_rmaps_celltype_condition, "no correction", f'{time}')
        try_dir(dir_rmaps_out)
        if constitutive_correction and func_cc_toggle:
            ccorr_data = df_ccorr_data[f'{celltype}_{condition}_{time}'].dropna().tolist()
            dir_rmaps_out_ccorr = Path(dir_rmaps_celltype_condition, "with correction", f'{time}')
            try_dir(dir_rmaps_out_ccorr)
            
        print(f'Ratio-mapping {celltype}, {condition}, {time}')
        
        # use sampled specimens (wells) if applicable, otherwise
        # generate ratio maps for whole data set (can take a very
        # long time)
        if samples_assigned:
            wells = get_well_list(time_flist_samples)
        else:
            wells = get_well_list(time_flist)
            
        # generate ratio map, correct it if constitutive correction
        # is enabled
        for num_well, denom_well in wells:
            try:
                df_ratiomap = get_ratiomap(num_well, denom_well)
                if constitutive_correction and func_cc_toggle:
                    df_ratiomap_ccorr = constitutive_correct_ratiomap(df_ratiomap, ccorr_data)
                    norm_df_ratiomap_ccorr = df_ratiomap_ccorr / reference_df_ccorr.min().min()
                    write_ratiomap(norm_df_ratiomap_ccorr, dir_rmaps_out_ccorr, norm_max_ccorr, newcmap, num_well, "with correction")
                norm_df_ratiomap = df_ratiomap / reference_df.min().min()
                write_ratiomap(norm_df_ratiomap, dir_rmaps_out, norm_max, newcmap, num_well, "no correction")
            except KeyboardInterrupt:
                raise KeyboardInterrupt()
            except:
                
                # if images for specific time points are missing,
                # user will be notified but the whole thing won't
                # crash
                print(f'Unable to generate a ratio map for {num_well}, {denom_well} - file(s) may be missing.')
                continue
                
    # generates a colorbar that corresponds to the scaling of the
    # colormap
    if constitutive_correction and func_cc_toggle:
        write_cbar(reference_df_ccorr.min().min(), reference_df_ccorr.max().max(), dir_rmaps_celltype_condition, newcmap, func_ratio_expression, f'{celltype} {condition}', "with correction")
    write_cbar(reference_df.min().min(), reference_df.max().max(), dir_rmaps_celltype_condition, newcmap, func_ratio_expression, f'{celltype} {condition}', "no correction")
```

```python
def make_rmaps_compare_conditions():
    # creates ratio maps with colour scaling based on the data for a
    # given cell line. colour scales across all time points and
    # conditions. useful for comparing between conditions and times,
    # but not between cell lines.
    
    celltype_info_sets = [x for x in rmaps_reference_info if x[0] == celltype]
    func_rmaps_cmap = celltype_info_sets[0][4]
    func_num_fluor = celltype_info_sets[0][5]
    func_denom_fluor = celltype_info_sets[0][6]
    func_ratio_expression = f'{func_num_fluor}/{func_denom_fluor}'
    func_cc_toggle = celltype_info_sets[0][-1]
    dir_rmaps_celltype = Path(dir_output, f'ratio-mapped images by celltype - scaled for all conditions, all time points ({rmaps_cmap.split(".")[-1]})', celltype)
    try_dir(dir_rmaps_celltype)
    
    # ratio map colormap is taken but lowest value replaced with
    # black to force contrast with background. for this reason, it is
    # generally best not to use colormaps with black on the low end
    # if low ratios are expected near the surface of the spheroid
    # (e.g., for a hypoxia biosensor)
    colors = plt.colormaps[func_rmaps_cmap](np.linspace(0, 1, 2**16))
    black = np.array([0, 0, 0, 1])
    colors[:1, :] = black
    newcmap = ListedColormap(colors)
    
    # mins and maxes are aggregated across conditions for a given
    # cell type, then the overall min and max are determined to set
    # color scaling
    mins = []
    maxes = []
    mins_ccorr = []
    maxes_ccorr = []
    
    for condition in conditions:
        info = [x for x in celltype_info_sets if x[1] == condition]
        if not info:
            continue
        reference_df = info[0][2]
        mins.append(reference_df.min().min())
        maxes.append(reference_df.max().max())
        if constitutive_correction and func_cc_toggle:
            reference_df_ccorr = info[0][7]
            mins_ccorr.append(reference_df_ccorr.min().min())
            maxes_ccorr.append(reference_df_ccorr.max().max())
            
    celltype_min = min(mins)
    celltype_max = max(maxes)
    norm_celltype_max = celltype_max / celltype_min
    
    # if correction is applied to ratio profiles, it is also
    # applied to ratio maps
    if constitutive_correction and func_cc_toggle:
        celltype_min_ccorr = min(mins_ccorr)
        celltype_max_ccorr = max(maxes_ccorr)
        norm_celltype_max_ccorr = celltype_max_ccorr / celltype_min_ccorr
        
    print(f'Ratio-mapping images (scaling for {celltype}, all conditions), colormap = {func_rmaps_cmap}...')
    
    for condition in conditions:
        info = [x for x in celltype_info_sets if x[1] == condition]
        if not info:
            continue
            
        condition_flist = [item for item in celltype_flist if item[1] == condition]
        if not condition_flist:
            continue
            
        if constitutive_correction and func_cc_toggle:
            df_ccorr_data = df_ccorr_data = info[0][8]
        
        # if using rmaps_sample, randomly samples the data set and ratio
        # maps the same speciments at each time point (nice for
        # representative images)
        samples_assigned = False
        for time in times:
            time_flist = [item for item in condition_flist if (item[6] == time) and (item[2] == func_num_fluor)]
            if not time_flist:
                continue
                
            if rmaps_sample and not samples_assigned:
                if rmaps_sample_wells < len(time_flist):
                    time_flist_samples = random.sample(time_flist, rmaps_sample_wells)
                    samples_assigned = True
            elif rmaps_sample and samples_assigned:
                for x in time_flist_samples:
                    x[6] = time
                    
            dir_rmaps_out = Path(dir_rmaps_celltype, f'{celltype}, {condition}', "no correction", f'{time}')
            try_dir(dir_rmaps_out)
            if constitutive_correction and func_cc_toggle:
                ccorr_data = df_ccorr_data[f'{celltype}_{condition}_{time}'].dropna().tolist()
                dir_rmaps_out_ccorr = Path(dir_rmaps_celltype, f'{celltype}, {condition}', "with correction", f'{time}')
                try_dir(dir_rmaps_out_ccorr)
                
            print(f'Ratio-mapping {celltype}, {condition}, {time}')
            
            # use sampled specimens (wells) if applicable, otherwise
            # generate ratio maps for whole data set (can take a very
            # long time)
            if samples_assigned:
                wells = get_well_list(time_flist_samples)
            else:
                wells = get_well_list(time_flist)
                
            # generate ratio map, correct it if constitutive correction
            # is enabled
            for num_well, denom_well in wells:
                try:
                    df_ratiomap = get_ratiomap(num_well, denom_well)
                    if constitutive_correction and func_cc_toggle:
                        df_ratiomap_ccorr = constitutive_correct_ratiomap(df_ratiomap, ccorr_data)
                        norm_df_ratiomap_ccorr = df_ratiomap_ccorr / celltype_min_ccorr
                        write_ratiomap(norm_df_ratiomap_ccorr, dir_rmaps_out_ccorr, norm_celltype_max_ccorr, newcmap, num_well, "with correction")
                    norm_df_ratiomap = df_ratiomap / celltype_min
                    write_ratiomap(norm_df_ratiomap, dir_rmaps_out, norm_celltype_max, newcmap, num_well, "no correction")
                except KeyboardInterrupt:
                    raise KeyboardInterrupt()
                except:
                    # if images for specific time points are missing,
                    # user will be notified by the whole thing won't
                    # crash
                    print(f'Unable to generate a ratio map for {num_well}, {denom_well} - file(s) may be missing.')
                    continue
                    
    # generates a colorbar that corresponds to the scaling of the
    # colormap
    if constitutive_correction and func_cc_toggle:
        write_cbar(celltype_min_ccorr, celltype_max_ccorr, dir_rmaps_celltype, newcmap, func_ratio_expression, f'{celltype}', "with correction")
    write_cbar(celltype_min, celltype_max, dir_rmaps_celltype, newcmap, func_ratio_expression, f'{celltype}', "no correction")
```

```python
def make_rmaps_compare_celltypes():
    # creates ratio maps with colour scaling based on the data for a
    # all cell lines expressing a given biosensor/biosensor. colour
    # scales across all time points, conditions, and applicable cell
    # lines. useful for comparing between cell lines, conditions, and
    # times
    
    biosensors = sorted(set([x[3] for x in rmaps_reference_info]), key = lambda v: v.upper())
    for biosensor in biosensors:
        dir_rmaps_biosensor = Path(dir_output, f'ratio-mapped images by biosensor - scaled for all celltypes, all conditions, all time points', f'{biosensor}')
        try_dir(dir_rmaps_biosensor)
        
        # identifies which cell lines are applicable for a given
        # biosensor/biosensor
        biosensor_info_sets = [x for x in rmaps_reference_info if x[3] == biosensor]
        biosensor_celltypes = sorted(set([x[0] for x in biosensor_info_sets]))
        
        func_rmaps_cmap = biosensor_info_sets[0][4]
        func_num_fluor = biosensor_info_sets[0][5]
        func_denom_fluor = biosensor_info_sets[0][6]
        func_ratio_expression = f'{func_num_fluor}/{func_denom_fluor}'
        func_cc_toggle = biosensor_info_sets[0][-1]
        
        # ratio map colormap is taken but lowest value replaced with
        # black to force contrast with background. for this reason, it is
        # generally best not to use colormaps with black on the low end
        # if low ratios are expected near the surface of the spheroid
        # (e.g., for a hypoxia biosensor)
        colors = plt.colormaps[func_rmaps_cmap](np.linspace(0, 1, 2**16))
        black = np.array([0, 0, 0, 1])
        colors[:1, :] = black
        newcmap = ListedColormap(colors)
        
        # mins and maxes are aggregated across conditions for a given
        # biosensor/biosensor, then the overall min and max are
        # determined to set color scaling. if correction is applied
        # to ratio profiles, it is also applied to ratio maps
        mins = []
        maxes = []
        mins_ccorr = []
        maxes_ccorr = []
        
        for celltype in biosensor_celltypes:
            for condition in conditions:
                info = [x for x in biosensor_info_sets if x[0] == celltype and x[1] == condition]
                if not info:
                    continue
                reference_df = info[0][2]
                mins.append(reference_df.min().min())
                maxes.append(reference_df.max().max())
                if constitutive_correction and func_cc_toggle:
                    reference_df_ccorr = info[0][7]
                    mins_ccorr.append(reference_df_ccorr.min().min())
                    maxes_ccorr.append(reference_df_ccorr.max().max())
                                
        biosensor_min = min(mins)
        biosensor_max = max(maxes)
        norm_biosensor_max = biosensor_max / biosensor_min
        if constitutive_correction and func_cc_toggle:
            biosensor_min_ccorr = min(mins_ccorr)
            biosensor_max_ccorr = max(maxes_ccorr)
            norm_biosensor_max_ccorr = biosensor_max_ccorr / biosensor_min_ccorr
            
        print(f'Ratio-mapping images (scaling for {biosensor}, all celltypes), colormap = {func_rmaps_cmap}')
        
        for celltype in biosensor_celltypes:
            celltype_flist = [item for item in file_details if item[0] == celltype]
            if not celltype_flist:
                continue
                
            for condition in conditions:
                info = [x for x in biosensor_info_sets if x[0] == celltype and x[1] == condition]
                if not info:
                    continue
                    
                condition_flist = [item for item in celltype_flist if item[1] == condition]
                if not condition_flist:
                    continue
                    
                if constitutive_correction and func_cc_toggle:
                    df_ccorr_data = info[0][8]
                    
                samples_assigned = False
                
                for time in times:
                    time_flist = [item for item in condition_flist if (item[6] == time) and (item[2] == func_num_fluor)]
                    if not time_flist:
                        continue
                        
                    if rmaps_sample and not samples_assigned:
                        if rmaps_sample_wells < len(time_flist):
                            time_flist_samples = random.sample(time_flist, rmaps_sample_wells)
                            samples_assigned = True
                    elif rmaps_sample and samples_assigned:
                        for x in time_flist_samples:
                            x[6] = time
                            
                    dir_rmaps_out = Path(dir_rmaps_biosensor, celltype, f'{celltype} {condition}', "no correction", f'{time}')
                    try_dir(dir_rmaps_out)
                    if constitutive_correction and func_cc_toggle:
                        ccorr_data = df_ccorr_data[f'{celltype}_{condition}_{time}'].dropna().tolist()
                        dir_rmaps_out_ccorr = Path(dir_rmaps_biosensor, celltype, f'{celltype} {condition}', "with correction", f'{time}')
                        try_dir(dir_rmaps_out_ccorr)
                        
                    print(f'Ratio-mapping {celltype}, {condition}, {time}')
                    
                    if samples_assigned:
                        wells = get_well_list(time_flist_samples)
                    else:
                        wells = get_well_list(time_flist)
                        
                    for num_well, denom_well in wells:
                        try:
                            df_ratiomap = get_ratiomap(num_well, denom_well)
                            if constitutive_correction and func_cc_toggle:
                                df_ratiomap_ccorr = constitutive_correct_ratiomap(df_ratiomap, ccorr_data)
                                norm_df_ratiomap_ccorr = df_ratiomap_ccorr / biosensor_min_ccorr
                                write_ratiomap(norm_df_ratiomap_ccorr, dir_rmaps_out_ccorr, norm_biosensor_max_ccorr, newcmap, num_well, "with correction")
                            norm_df_ratiomap = df_ratiomap / biosensor_min
                            write_ratiomap(norm_df_ratiomap, dir_rmaps_out, norm_biosensor_max, newcmap, num_well, "no correction")
                        except KeyboardInterrupt:
                            raise KeyboardInterrupt()
                        except:
                            print(f'Unable to generate a ratio map for {num_well}, {denom_well} - file(s) may be missing.')
                            continue
                            
        # generates a colorbar that corresponds to the scaling of the
        # colormap
        if constitutive_correction and func_cc_toggle:
            write_cbar(biosensor_min_ccorr, biosensor_max_ccorr, dir_rmaps_biosensor, newcmap, func_ratio_expression, f'{biosensor}', "with correction")
        write_cbar(biosensor_min, biosensor_max, dir_rmaps_biosensor, newcmap, func_ratio_expression, f'{biosensor}', "no correction")
```

```python
def mirror_df(*dfs):
    # mirrors a DataFrame
    # used for generating mirrored ratio profiles
    
    # create a flipped version, then concatenate
    mirrored_dfs = []
    for df in dfs:
        right_half_df = df.copy()
        left_half_df = df.copy().iloc[::-1].reset_index(drop = True)
        mirrored_df = pd.concat([left_half_df, right_half_df]).reset_index(drop = True)
        mirrored_dfs.append(mirrored_df)
    if len(mirrored_dfs) == 1:
        mirrored_dfs = mirrored_dfs[0]
    return(mirrored_dfs)
```

```python
def polygon_under_graph(x, y):
    # generates a list of coordinates to fill under when generating
    # each trace on a 3D graph. (x[0], 0.) and (x[-1], 0.) terms
    # force the graphs to fill down to the x axis at z = 0. from
    # matplotlib documentation:
    # https://matplotlib.org/stable/gallery/mplot3d/polys3d.html
    
    return [(x[0], 0.), *zip(x, y), (x[-1], 0.)]
```

```python
def process_image_intensity(df):
    # gaussian blur, background subtraction, centroid determination
    
    # minor blur and background subtraction
    df_gaussian = pd.DataFrame(gaussian_filter(df, sigma = 1))
    bg = np.percentile(df_gaussian, bg_percentile)
    df_gaussian_bgsub = (df_gaussian - bg).clip(lower = 0)
    
    # create a mask for the foreground, accounts for small numbers
    # of extremely bright pixels
    sat = np.percentile(df_gaussian_bgsub, 98)
    df_gaussian_bgsub_robust_mask = df_gaussian_bgsub.clip(upper = sat)
    thresh = skimage.filters.threshold_otsu(df_gaussian_bgsub_robust_mask.values) * 0.5
    df_masked = df_gaussian_bgsub.where(df_gaussian_bgsub_robust_mask > thresh, other = 0)
    
    # determines foreground coordinates and calculates centroid
    # coordinate
    foreground_coords = np.argwhere(df_masked > 0)
    centroid = np.mean(foreground_coords, axis = 0, dtype = int)
    return(df_masked, centroid, foreground_coords)
```

```python
def process_image_ratio(df_num, df_denom):
    
    # gaussian blur, background subtraction, centroid determination
    
    # minor blur and background subtraction
    df_num_gaussian = pd.DataFrame(gaussian_filter(df_num, sigma = 1))
    bg_num = np.percentile(df_num_gaussian, bg_percentile)
    df_num_gaussian_bgsub = (df_num_gaussian - bg_num).clip(lower = 0)
    df_denom_gaussian = pd.DataFrame(gaussian_filter(df_denom, sigma = 1))
    bg_denom = np.percentile(df_denom_gaussian, bg_percentile)
    df_denom_gaussian_bgsub = (df_denom_gaussian - bg_denom).clip(lower = 0)
    
    # create masks for the foreground of each channel, accounts for small numbers
    # of extremely bright pixels
    sat_num = np.percentile(df_num_gaussian_bgsub, 98)
    df_num_gaussian_bgsub_robust_mask = df_num_gaussian_bgsub.clip(upper = sat_num)
    thresh_num = skimage.filters.threshold_otsu(df_num_gaussian_bgsub_robust_mask.values) * 0.5
    sat_denom = np.percentile(df_denom_gaussian_bgsub, 98)
    df_denom_gaussian_bgsub_robust_mask = df_denom_gaussian_bgsub.clip(upper = sat_denom)
    thresh_denom = skimage.filters.threshold_otsu(df_denom_gaussian_bgsub_robust_mask.values) * 0.5

    # calculate SNR as threshold / background
    SNR_num = thresh_num / bg_num
    SNR_denom = thresh_denom / bg_denom
    
    # compare SNRs to determine which channel is better to use for
    # masking to identify the foreground
    if SNR_num > SNR_denom:
        df_num_masked = df_num_gaussian_bgsub.where(df_num_gaussian_bgsub_robust_mask > thresh_num, other = 0)
        df_denom_masked = df_denom_gaussian_bgsub.where(df_num_gaussian_bgsub_robust_mask > thresh_num, other = 0)
    elif SNR_num < SNR_denom:
        df_num_masked = df_num_gaussian_bgsub.where(df_denom_gaussian_bgsub_robust_mask > thresh_denom, other = 0)
        df_denom_masked = df_denom_gaussian_bgsub.where(df_denom_gaussian_bgsub_robust_mask > thresh_denom, other = 0)
    elif SNR_num == SNR_denom:
        # if SNRs are equal, uses a combined mask
        df_num_masked = df_num_gaussian_bgsub.where((df_num_gaussian_bgsub_robust_mask > thresh_num) | (df_denom_gaussian_bgsub_robust_mask > thresh_denom), other = 0)
        df_denom_masked = df_denom_gaussian_bgsub.where((df_num_gaussian_bgsub_robust_mask > thresh_num) | (df_denom_gaussian_bgsub_robust_mask > thresh_denom), other = 0)
    
    # determines foreground coordinates and calculates centroid coordinate
    # foreground_coords = np.argwhere((df_num_gaussian_bgsub_robust_mask > thresh_num) | (df_denom_gaussian_bgsub_robust_mask > thresh_denom))
    foreground_coords = np.argwhere(df_num_masked > 0)
    centroid = np.mean(foreground_coords, axis = 0, dtype = int)
    return(df_num_masked, df_denom_masked, centroid, foreground_coords)
```

```python
def process_wells_intensity(wells):
    # central function for coordinating data extraction and
    # aggregating across a subset of paired images
    
    df_intensities = initialize_df()
    
    # extract data from images using get_intensities and deposit into
    # DataFrame
    for well in wells:
        try:
            intensities = get_intensities_intensity(well).reset_index(drop = True)
            label = f'{Path(well).parts[-1]}_intensities'
            df_intensities[label] = intensities
        except KeyboardInterrupt:
            raise KeyboardInterrupt()
        except:
            continue
    df_intensities = cleanup_df(df_intensities)
    return(df_intensities)
```

```python
def process_wells_ratio(wells):
    # central function for coordinating data extraction and
    # aggregating across a subset of paired images
    
    df_num_intensities = initialize_df()
    df_denom_intensities = initialize_df()
    
    # extract data from paired images using get_intensities and
    # deposit into DataFrames
    for num_well, denom_well in wells:
        try:
            num_intensities, denom_intensities = get_intensities_ratio(num_well, denom_well)
            num_label = f'{Path(num_well).stem}_intensities'
            denom_label = f'{Path(denom_well).stem}_intensities'
            df_num_intensities[num_label] = num_intensities
            df_denom_intensities[denom_label] = denom_intensities
        except KeyboardInterrupt:
            raise KeyboardInterrupt()
        except:
            continue
    df_num_intensities, df_denom_intensities = cleanup_df(df_num_intensities, df_denom_intensities)
    return(df_num_intensities, df_denom_intensities)
```

```python
def resample_df(*dfs, how = "asym"):
    # equalizes the length of all profiles contained in a DataFrame
    # via resample_list_to. necessary for determining mean.
    # resampling does not change curve shape, but adds or removes
    # values. resampling sets all profiles to the average profile
    # length for the DataFrame
    
    resampled_dfs = []
    for df_raw in dfs:
        df = df_raw.copy()
        df = cleanup_df(df)
        if df.empty:
            resampled_dfs.append(df)
            continue
            
        # separate each profile, get lengths, get average length
        col_list = [df[x].dropna().tolist() for x in df.columns]
        col_lengths = [len(x) for x in col_list]
        if not col_lengths:
            resampled_dfs.append(df)
            continue
        col_lengths_avg = int(np.average(col_lengths))
        
        # resample each profile to the average length and create a
        # new DataFrame from the resampled profiles
        resampled_cols = []
        for col in col_list:
            resampled_col = resample_list_to(col, col_lengths_avg, how)
            resampled_cols.append(resampled_col)
        df_resampled = pd.DataFrame(resampled_cols).T
        df_resampled.columns = df.columns
        resampled_dfs.append(df_resampled)
    if len(resampled_dfs) == 1:
        resampled_dfs = resampled_dfs[0]
    return(resampled_dfs)
```

```python
def resample_list_to(list_values, n, how = "asym"):
    # adds or removes values from a profile without changing the
    # shape of the curve
    
    # if the profile needs to be shortened, remove the necessary
    # number of values at evenly spaced intervals from within the
    # original list of values
    if len(list_values) > n:
        resampled = [list_values[y] for y in [int(x) for x in np.linspace(0, len(list_values) - 1, n)]]
        
    # if the profile needs to be lengthened, interpolate
    elif len(list_values) < n:
        initx = np.linspace(0, len(list_values) - 1, len(list_values))
        finx = np.linspace(0, len(list_values) - 1, n)
        resampled = np.interp(finx, initx, list_values)
    elif len(list_values) == n:
        resampled = list_values
        
    # mostly a relic from original versions, but takes a mirrored
    # average of a profile
    if how == "sym": 
        forward = pd.Series(resampled)
        reverse = forward.copy()[::-1].reset_index(drop = True)
        resampled = list((forward + reverse) / 2)
    return(resampled)
```

```python
def roll_df(window, *dfs):
    # takes a rolling average of each profile contained within a
    # DataFrame via roll_series
    
    rolling_dfs = []
    for df in dfs:
        df_rolling = df.copy()
        for col in df.columns:
            df_rolling[col] = roll_series(window, df[col])
        rolling_dfs.append(df_rolling)
    if len(rolling_dfs) == 1:
        rolling_dfs = rolling_dfs[0]
    return(rolling_dfs)
```

```python
def roll_series(window, ser):
    # takes a rolling average of a series (profile) based on user
    # input. functions very similarly to trim_series
    
    values = ser.dropna().tolist()
    
    # if user input for roll is less than 0.5, treat as a fraction
    if 0 < window <= 0.5:
        roll_no = int(len(values) * window)
        
    # if user input for roll is between 0.5 and 1.0, "truncate" to
    # 0.5
    elif 0.5 < window < 1:
        roll_no = int(len(values) * 0.5)
        
    # if user input for roll exceeds 1.0, treat as a number of values
    # unless roll is over half the number of values in the profile,
    # in which case it gets "truncated" to half (similar to above)
    elif window >= 1:
        roll_no = int(roll)
        if window > len(values) / 2:
            roll_no = int(len(values) / 2)
            
    # handles things like negative number input
    else:
        roll_no = 1
        
    # if the value ends up being less than 1, set it to 1 (can't take
    # rolling average with a window size of 0)
    if roll_no < 1:
        roll_no = 1
        
    rolled_values = ser.rolling(roll_no, center = True).mean()
    return(rolled_values)
```

```python
def trim_df(trim, direction, *dfs):
    # trims each profile contained within a DataFrame via trim_series
    
    trimmed_dfs = []
    for df in dfs:
        df_trimmed = df.copy()
        for col in df.columns:
            df_trimmed[col] = trim_series(trim, direction, df[col])
        trimmed_dfs.append(df_trimmed)
    if len(trimmed_dfs) == 1:
        trimmed_dfs = trimmed_dfs[0]
    return(trimmed_dfs)
```

```python
def trim_series(trim, direction, ser):
    # trims a series (profile) based on user input. functions very
    # similarly to roll_series
    
    values = ser.dropna().tolist()
    
    # if user input for trim is less than 0.5, treat as a fraction
    if 0 <= trim <= 0.5:
        trim_no = int(len(values) * trim)
        
    # if user input for trim is between 0.5 and 1.0, "truncate" to
    # 0.5
    elif 0.5 < trim < 1:
        trim_no = int(len(values) * 0.5)
        
    # if user input for trim exceeds 1.0, treat as a number of values
    # unless trim is over half the number of values in the profile,
    # in which case it gets "truncated" to half (similar to above)
    elif trim >= 1:
        trim_no = int(trim)
        if trim > len(values) / 2:
            trim_no = int(len(values) / 2)
            
    # handles things like negative number input
    else:
        trim_no = 0
        
    # does the trimming, considers orientation of the profile
    if direction == "centre-out":
        trimmed_values = values[0: len(values) - trim_no]
    elif direction == "surface-in":
        trimmed_values = values[trim_no: len(values)]
        nans = [np.nan] * trim_no
        trimmed_values = nans + trimmed_values
    return(pd.Series(trimmed_values))
```

```python
def try_dir(some_dir):
    # creates whatever directory, but checks whether or not it exists
    # first. mostly implemented as a debugging tool
    
    if not os.path.exists(some_dir):
        os.makedirs(some_dir)
```

```python
def write_all_times_intensity_data(dfs):
    # central function for writing outputs. iterates over user
    # selected settings and generates all combinations, all
    # relevant formats, etc. functions similarly to
    # write_timepoint_data
    
    # handles centre-out vs. surface-in
    for direction in directions:
        df_to_write = dfs[0].copy()
        df_error_to_write = dfs[1].copy()
        dir_direction = Path(dir_celltype_condition_fluor, direction)
        try_dir(dir_direction)
        dir_graphs = Path(dir_direction, f'time course graphs ({graphs_cmap.split(".")[-1]}) - {trimming}, {smoothing}')
        out_name = f'{celltype} {condition} {fluor} (mean) - {direction} intensity profiles and error.xlsx'
        direction_sheets = []
        if direction == "surface-in":
            df_to_write = flip_df(df_to_write)
            df_error_to_write = flip_df(df_error_to_write)
            
        # handles different data formats (raw, relative, percentage)
        # and deals with the required transformations for each
        for n, data_format in enumerate(data_formats):
            df_output = df_to_write.copy()
            df_output_error = df_error_to_write.copy()
            if data_format[0].endswith("rel to 1 (overall)"):
                df_output = df_to_write / df_to_write.min().min()
                df_output_error = df_error_to_write / df_to_write.min().min()
            elif data_format[0].endswith("rel to 1 (per time)"):
                df_output = df_to_write / df_to_write.min()
                df_output_error = df_error_to_write / df_to_write.min()
            elif data_format[0].endswith("pct of max (overall)"):
                df_output = (df_to_write / df_to_write.max().max()) * 100
                df_output_error = (df_error_to_write / df_to_write.max().max()) * 100
            elif data_format[0].endswith("pct of max (per time)"):
                df_output = (df_to_write / df_to_write.max()) * 100
                df_output_error = (df_error_to_write / df_to_write.max()) * 100
            direction_sheets.append([df_output, n, data_format, False])
            direction_sheets.append([df_output_error, n, data_format, True])
            make_graphs_2d_intensity((df_output, df_output_error), dir_graphs, direction, data_format)
            make_graphs_3d_intensity((df_output, df_output_error), dir_graphs, direction, data_format)
            
            # only makes mirrored graphs for centre-out data
            if direction == "centre-out" and mirrored_graphs:
                df_output_mirrored, df_output_error_mirrored = mirror_df(df_output, df_output_error)
                dir_graphs_mirrored = Path(dir_celltype_condition_fluor, f'mirrored', f'time course graphs ({graphs_cmap.split(".")[-1]}) - {trimming}, {smoothing}')
                make_graphs_2d_intensity((df_output_mirrored, df_output_error_mirrored), dir_graphs_mirrored, direction, data_format, mirrored = True)
                make_graphs_3d_intensity((df_output_mirrored, df_output_error_mirrored), dir_graphs_mirrored, direction, data_format, mirrored = True)
                
        # creates a workbook with sheets containing the numerical
        # data that has been extracted. sheets for values and error
        # are separate
        with pd.ExcelWriter(Path(dir_direction, out_name)) as writer:
            for item in direction_sheets:
                sheet = item[0]
                z = item[1]
                data_format = item[2]
                error = item[3]
                if error:
                    target_sheet = f'error{z + 1} {data_format[0].split(") - ")[1]}'[:31]
                    sheet.to_excel(writer, sheet_name = target_sheet, index = False)
                else:
                    target_sheet = f'profs{z + 1} {data_format[0].split(") - ")[1]}'[:31]
                    sheet.to_excel(writer, sheet_name = target_sheet, index = False)
```

```python
def write_cbar(data_min, data_max, dir_target, some_cmap, func_ratio_expression, cbar_title, ccorr_status):
    # function to generate colorbar, used in generation of ratio maps
    
    cbar_title_parts = cbar_title.split(" ")
    cbar_title_label = f'\n'.join(cbar_title_parts)
    cbar_formats = sorted(set([x[0].split(") - ")[-1].split(" (")[0] for x in data_formats]))
    for cbar_format in cbar_formats:
        if cbar_format == "raw":
            cbar_min, cbar_max = data_min, data_max
            cbar_label = f'{func_ratio_expression} ratio'
        elif cbar_format == "rel to 1":
            cbar_min, cbar_max = 1, data_max / data_min
            cbar_label = f'Rel. {func_ratio_expression} ratio'
        elif cbar_format == "pct of max":
            cbar_min, cbar_max = (data_min / data_max) * 100, 100
            cbar_label = f'Rel. {func_ratio_expression} ratio\n(% of max)'
            
        # prevents visual gaps between colorbar and border at
        # extremes; these presumably happen due to rounding
        norm = Normalize(vmin = cbar_min, vmax = cbar_max)
        tick_min = math.floor(cbar_min / 0.000001) * 0.000001
        if tick_min == 0:
            tick_min = cbar_min
        tick_max = math.ceil(cbar_max / 0.000001) * 0.000001
        if tick_max == 0:
            tick_max = cbar_max
            
        if cbar_format == "pct of max":
            tick_max = 100.0
            
        # set up ticks. tried this a lot of different ways, but
        # always ended up with weird behaviour. cleanest and easiest
        # was to just stick with min, max, and half-way point. labels
        # have two decimal places, can be adjusted as necessary
        ticks = np.linspace(tick_min, tick_max, 3)
        tick_labels = [str(np.round(x, decimals = 2)) for x in ticks]
        
        # colorbar orientation etc.
        plt.figure(figsize = (4.592, 1.325))
        cbar = plt.colorbar(plt.cm.ScalarMappable(norm = norm, cmap = some_cmap), ax = plt.gca(), fraction = 1, aspect = 10, orientation = 'horizontal')
        # plt.figure(figsize = (1.325, 4.592))
        # cbar = plt.colorbar(plt.cm.ScalarMappable(norm = norm, cmap = some_cmap), ax = plt.gca(), fraction = 1, aspect = 10, orientation = 'vertical')
        cbar.outline.set_linewidth(2)
        cbar.set_label(cbar_label, size = 22, labelpad = 3, fontweight = 500)
        cbar.ax.tick_params(labelsize = 20, length = 8, width = 2, pad = 3)
        cbar.set_ticks(ticks, labels = tick_labels)
        cbar.ax.yaxis.set_label_position('left')
        cbar.ax.set_title(cbar_title_label, size = 20, pad = 25, fontweight = 500)
        plt.axis('off')
        plt.savefig(Path(dir_target, f'{cbar_title} colormap ({ccorr_status}), {cbar_format}.png'), dpi = 300, bbox_inches = 'tight')
        plt.close()
```

```python
def write_ccorr_data(df):
    # outputs and plots (if desired) the constitutive correction
    # ratio data. essentially, what the "expected" ratio profiles
    # would look like based on the input data
    
    ccorr_status = "constitutive correction data"
    data_format = ("rel to 1 (per time)", "relative")
    df_to_write = df.copy()
    df_error_to_write = df_to_write * 0
    dir_ccorr = Path(dir_celltype_condition, f'constitutive correction data (centre-out only) - no trimming, no smoothing')
    try_dir(dir_ccorr)
    out_name = f'{celltype} {condition} - centre-out {ccorr_status} ({data_format[0]}).xlsx'
    
    # outputs a workbook with numerical values. error values are not
    # generated or considered for constitutive correction data
    with pd.ExcelWriter(Path(dir_ccorr, out_name)) as writer:
        df_to_write.to_excel(writer, index = False)
    make_graphs_2d_ratio((df_to_write, df_error_to_write), dir_ccorr, "centre-out", data_format, ccorr_status)
    make_graphs_3d_ratio((df_to_write, df_error_to_write), dir_ccorr, "centre-out", data_format, ccorr_status)
    if mirrored_graphs:
        df_to_write_mirrored, df_error_to_write_mirrored = mirror_df(df_to_write, df_error_to_write)
        make_graphs_2d_ratio((df_to_write_mirrored, df_error_to_write_mirrored), dir_ccorr, "centre-out", data_format, ccorr_status, mirrored = True)
        make_graphs_3d_ratio((df_to_write_mirrored, df_error_to_write_mirrored), dir_ccorr, "centre-out", data_format, ccorr_status, mirrored = True)
```

```python
def write_ratiomap(df, dir_target, some_max, some_cmap, num_well, ccorr_status):
    # scales the color for a given ratio map and writes the output
    # image

    # scales color in the image based on values close to the extremes
    # but not quite at them. ratios can be finnicky, especially if
    # background subtraction leaves values that are very close to 0
    # in either the numerator or denominator image
    image_min = np.percentile(df.values[df > 0], 1)
    if image_min > 1:
        vmin = 1
    else:
        vmin = image_min * 0.99
    image_max = np.percentile(df.values[df > 0], 99)

    # assumes that ridiculously high (>50-fold the average max)
    # values are artifactual and ignores them
    if image_max <= some_max or image_max >= (50 * some_max):
        vmax = some_max
    else:
        vmax = image_max

    plt.figure(figsize = (6.097, 4.503))
    plt.axis('off')
    plt.imshow(df, cmap = some_cmap, vmin = vmin, vmax = vmax, interpolation = 'lanczos')
    new_name = f'{str(Path(num_well).stem).replace(f'_{num_fluor}_', "_mapped_")}.png'
    plt.savefig(Path(dir_target, new_name), dpi = 300, bbox_inches = 'tight', pad_inches = 0)
    plt.close()
```

```python
def write_timepoint_intensity_data(data, timepoint_means_extremes, regional_timepoint_means_extremes):
    # central function for writing outputs. iterates over user
    # selected settings and generates all combinations, all
    # relevant formats, etc. functions similarly to
    # write_all_times_data, but outputs profiles on a per-replicate
    # basis, versus the mean and error. also deals with regional data
    # (i.e., mean and error of segmented profiles)
    
    # handles centre-out vs. surface-in
    for direction in directions:
        dir_direction = Path(dir_celltype_condition_fluor, direction)
        try_dir(dir_direction)
        out_name = f'{celltype} {condition} {fluor} (replicates) - {direction} intensity profiles and regional data.xlsx'
        direction_sheets = []
        
        # handles different data formats (raw, relative, percentage)
        # and deals with the required transformations for each
        for n, data_format in enumerate(data_formats):
            single_format_all_profiles = []
            single_format_all_regional = []
            for item in data:
                df_profiles_to_write = item[0].copy()
                df_regional_to_write = item[1].copy()
                if direction == "surface-in":
                    df_profiles_to_write = flip_df(df_profiles_to_write)
                    df_regional_to_write = flip_df(df_regional_to_write)
                df_output = df_profiles_to_write.copy()
                df_regional_output = df_regional_to_write.copy()
                if data_format[0].endswith("rel to 1 (overall)"):
                    df_output = df_profiles_to_write / min(timepoint_means_extremes)
                    df_regional_output = df_regional_to_write / min(regional_timepoint_means_extremes)
                elif data_format[0].endswith("rel to 1 (per time)"):
                    df_output = df_profiles_to_write / df_profiles_to_write.mean(axis = 1).min()
                    df_regional_output = df_regional_to_write / df_regional_to_write.mean(axis = 1).min()
                elif data_format[0].endswith("pct of max (overall)"):
                    df_output = (df_profiles_to_write / max(timepoint_means_extremes)) * 100
                    df_regional_output = (df_regional_to_write / max(regional_timepoint_means_extremes)) * 100
                elif data_format[0].endswith("pct of max (per time)"):
                    df_output = (df_profiles_to_write / df_profiles_to_write.mean(axis = 1).max()) * 100
                    df_regional_output = (df_regional_to_write / df_regional_to_write.mean(axis = 1).max()) * 100
                    
                # all the individual time point data (per replicate)
                # is concatenated into a single sheet (for each data
                # format)
                df_output[''] = pd.Series([np.nan] * len(df_output))
                df_regional_output[''] = pd.Series([np.nan] * len(df_regional_output))
                single_format_all_profiles.append(df_output.copy())
                single_format_all_regional.append(df_regional_output.copy().T)
            profiles_format_sheet = pd.concat(single_format_all_profiles, axis = 1).T
            regional_format_sheet = pd.concat(single_format_all_regional)
            direction_sheets.append([profiles_format_sheet, n, data_format, False])
            direction_sheets.append([regional_format_sheet, n, data_format, True])
            
        # creates a workbook with sheets containing the numerical
        # data (per replicate) that has been extracted from each time
        # point. sheets for profiles and error are separate from
        # sheets with regional data
        with pd.ExcelWriter(Path(dir_direction, out_name)) as writer:
            for item in direction_sheets:
                sheet = item[0]
                z = item[1]
                data_format = item[2]
                regional = item[3]
                if regional:
                    target_sheet = f'regions{z + 1} {data_format[0].split(") - ")[1]}'[:31]
                    sheet.to_excel(writer, sheet_name = target_sheet)
                else:
                    target_sheet = f'profs{z + 1} {data_format[0].split(") - ")[1]}'[:31]
                    sheet.to_excel(writer, sheet_name = target_sheet, header = False)
```

# FILE PARSE

```python
# parses all files within input directory of the user-specified file
# type

inputs = [str(path) for path in Path(dir_input).rglob(f'*.{filetype}')]

# filters out any hidden files
files = filter_hidden(inputs)

# extracts "metadata" embedded in file names
file_details = [Path(file_name).stem.split("_") for file_name in files]

# identifies unique entries at each position in file name for looping
# in subsequent steps. for clarification, "celltype" refers to the
# combined cell-biosensor term
celltypes = sorted(set([details[0] for details in file_details]), key = lambda v: v.upper())
conditions = sorted(set([details[1] for details in file_details]), key = lambda v: v.upper())
fluors = sorted(set([details[2] for details in file_details]), key = lambda v: v.upper())
replicates = sorted(set([details[3] for details in file_details]), key = lambda v: v.upper())
times = sorted(set([details[-1] for details in file_details]), key = lambda v: v.upper())

# converts __d__h__m to hours for interpolation during constitutive
# correction, if necessary
times_in_hours = [(24 * int(time.split("_")[-1].split("d")[0]) + int(time.split("d")[1].split("h")[0]) + (int(time.split("h")[1].split("m")[0]) / 60)) for time in times]
time_tuples = list(zip(times, times_in_hours))
```

# INTENSITY MODULE

```python
# iterates over all combinations of cell type, condition, etc. for
# all time points in the data set

if intensity_or_ratio == "intensity":
    
    print(f'\nSPoRTS intensity analysis initiated using {profile_or_comprehensive} method.\n')
    constitutive_control_dfs, analyzed_celltypes = [], []
    
    # begins looping over celltypes. in contrast to the ratio analysis,
    # any information encoded after a dash in the celltype term is not
    # considered in the analysis. thus, celltype is treated as anything
    # in the file name prior to the first underscore
    for celltype in celltypes:
        
        # filters to a working list of files matching the current
        # celltype, then filters down to paired images
        celltype_flist = [item for item in file_details if item[0] == celltype]
        
        # begins looping over conditions. filters working list of files
        # down to those matching the current condition
        for condition in conditions:
            condition_flist = [item for item in celltype_flist if item[1] == condition] 
            if not condition_flist:
                continue
                
            # begins looping over fluorophores. filters working list of
            # files down to those matching the current fluorophore
            for fluor in fluors:
                fluor_flist = [item for item in condition_flist if item[2] == fluor]
                if not fluor_flist:
                    continue
                    
                # if a colormap choice is provided in output preferences,
                # it will be used for graphing. if not, the default
                # colormap is used instead
                try:
                    graphs_cmap = [x[1] for x in fluor_settings if x[0] == fluor][0]
                except:
                    graphs_cmap = default_graphs_cmap
                analyzed_celltypes.append(f'{celltype}_{fluor}')
                
                # sets up all the structures necessary to hold the extracted
                # data
                df_all_times = initialize_df()
                df_all_times_error = initialize_df()
                relevant_times_in_hours, timepoint_data, timepoint_means_extremes, regional_timepoint_means_extremes = [], [], [], []
                
                # begins looping over time points. filters working list of
                # files down to those matching the current time point
                for time_tuple in time_tuples:
                    time = time_tuple[0]
                    time_hours = time_tuple[1]
                    time_flist = [item for item in fluor_flist if item[6] == time]
                    if not time_flist:
                        continue
                    print (f'Processing: {celltype}, {condition}, {fluor}, {time}') 
                    relevant_times_in_hours.append(time_hours)
                    
                    # begins looping over (experimental) replicates.
                    # filters working list of files down to thosematching
                    # the current replicate
                    df_all_replicates = initialize_df()
                    for replicate in replicates:
                        replicate_flist = [item for item in time_flist if item[3] == replicate]
                        if not replicate_flist:
                            continue
                            
                        # generates list of files from metadata and
                        # uses process_wells to extract data into DataFrame
                        wells = [f'{(Path(dir_input, "_".join(item)))}.tif' for item in replicate_flist]
                        df_intensities = process_wells_intensity(wells)
                        
                        if df_intensities.empty:
                            print(f'Unable to extract usable data from {time} {replicate}.')
                            continue
                            
                        # reports profile lengths for current replicate
                        # so user can monitor data extraction progress
                        rep_profile_lengths = [len(df_intensities[x].dropna().tolist()) for x in df_intensities.columns]
                        print(f'Profile length(s) for {replicate} (radius, {distance_unit}): {[int(x * dist_per_px) for x in rep_profile_lengths]}')
                        
                        # generates average profile for the current
                        # combination of celltype, condition, time point,
                        # and replicate (i.e., mean of technical replicates)
                        avg_profile = resample_df(df_intensities).mean(axis = 1)
                        
                        # collects the average profile for each replicate
                        # (for a given time point) in a DataFrame
                        df_all_replicates[f'{celltype}_{condition}_{time}_{replicate}'] = avg_profile
                        
                    df_all_replicates = cleanup_df(df_all_replicates)
                    if df_all_replicates.empty:
                        print(f'Unable to extract usable data for {time}.')
                        continue
                        
                    # resample the DataFrame to the average profile length
                    # to allow for mean profile to be calculated (for all
                    # replicates at a given time point), and to trim/roll/
                    # normalize
                    time_profile_lengths = [len(df_all_replicates[x].dropna().tolist()) for x in df_all_replicates.columns]
                    df_all_replicates = resample_df(df_all_replicates)
                    print(f'Average profile length(s) for {time} (radius, {distance_unit}): {[int(x * dist_per_px) for x in time_profile_lengths]} -- {int(len(df_all_replicates.index) * dist_per_px)} overall\n')
                    if trim_values:
                        df_all_replicates = trim_df(trim, "centre-out", df_all_replicates)
                    if roll_values:
                        df_all_replicates = roll_df(roll, df_all_replicates)
                    if replicate_normalization:
                        df_all_replicates = df_all_replicates / (df_all_replicates.mean() / df_all_replicates.mean().mean())
                        
                    # calculates values as required to normalize profile and
                    # regional time point data later on, and packages time
                    # point data for analysis later (need to know the overall
                    # min and max values for overall normalization vs. 1.0 or
                    # vs. 100%)
                    timepoint_means_extremes.append(df_all_replicates.mean(axis = 1).min())
                    timepoint_means_extremes.append(df_all_replicates.mean(axis = 1).max())
                    regional_df = get_regional_data(df_all_replicates)
                    regional_timepoint_means_extremes.append(regional_df.mean(axis = 1).min())
                    regional_timepoint_means_extremes.append(regional_df.mean(axis = 1).max())
                    timepoint_data.append((df_all_replicates, regional_df))
                    
                    # get mean and error across replicates for a given time
                    # point, and aggregate results in "all times" DataFrames
                    df_all_times[f'{celltype}_{condition}_{time}'] = df_all_replicates.mean(axis = 1)
                    if error_type in ("SE", "SEM"):
                        df_all_times_error[f'{celltype}_{condition}_{time}'] = df_all_replicates.sem(axis = 1)
                    elif error_type in ("SD", "STD"):
                        df_all_times_error[f'{celltype}_{condition}_{time}'] = df_all_replicates.std(axis = 1)
                        
                # once all time points are complete, output the data for: all
                # times, individual time points, any graphical options, and
                # an analysis summary
                print(f'Generating outputs for {celltype} {condition} {fluor}...\n')
                try_dir(dir_output)
                dir_celltype_condition_fluor = Path(dir_output, f'{celltype} {condition} {fluor}')
                try_dir(dir_celltype_condition_fluor)
                summary_intensity()
                write_timepoint_intensity_data(timepoint_data, timepoint_means_extremes, regional_timepoint_means_extremes)
                df_all_times, df_all_times_error = cleanup_df(df_all_times, df_all_times_error)
                if df_all_times_error.empty:
                    df_all_times_error = df_all_times * 0
                write_all_times_intensity_data((df_all_times, df_all_times_error))
                
                # set aside data that will be used for constitutive
                # correction
                normalization_data = df_all_times.copy() / df_all_times.min()
                normalization_data.columns = [f't (hours) = {item}' for item in relevant_times_in_hours]
                constitutive_control_dfs.append((f'{celltype}_{condition}_{fluor}', normalization_data))
                
    # once all celltypes have been analyzed, write constitutive
    # correction data (this file is what is included in subsequent
    # ratio analyses)
    analyzed_celltypes = str(analyzed_celltypes).translate({ord(c): None for c in "'"})[:184]
    with pd.ExcelWriter(Path(dir_output, f'Constitutive data {analyzed_celltypes}.xlsx')) as writer:
        for item in constitutive_control_dfs:
            item[1].to_excel(writer, sheet_name = item[0][:31], index = False)
            
    # output cell below updates user on analysis progression. begins
    # processing based on alphabetical order of celltype -->
    # condition, starting with the earliest time point. reports
    # profile lengths while analysis is underway
```

```python
# announces when the run is complete
if intensity_or_ratio == "intensity":
    now = datetime.now()
    print("\n**********ANALYSIS COMPLETE**********")
    print(now.strftime("%Y-%m-%d_%H%M%S"))
```

# RATIO MODULE

```python
# iterates over all combinations of cell type, condition, etc. for
# all time points in the data set

if intensity_or_ratio == "ratio":
    print(f'\nSPoRTS ratio analysis initiated using {profile_or_comprehensive} method.')
    rmaps_reference_info = []
    analyzed_celltypes = celltypes.copy()

    # begins looping over celltypes
    for celltype in celltypes:
        biosensor = None
        config = None
        num_fluor = None
        denom_fluor = None
        ratio_expression = None
        graphs_cmap = None
        rmaps_cmap = None

        # parses biosensor information, checks for unpaired files,
        # and assigns variables initialized above (i.e., biosensor,
        # config, etc.)
        celltype_flist = celltype_init(celltype, report_info = True)
        if not celltype_flist:
            print(f'\nRatio analysis could not be performed for {celltype}, since no paired files were found in the input directory.')
            analyzed_celltypes.remove(celltype)
            continue

        analyzed_celltypes = celltypes.copy()
        # begins looping over conditions. filters working list of files
        # down to those matching the current condition
        for condition in conditions:
            condition_flist = [item for item in celltype_flist if item[1] == condition]
            if not condition_flist:
                continue
                
            # sets up all the structures necessary to hold the extracted
            # data
            timepoint_data, timepoint_means_extremes, regional_timepoint_means_extremes = [], [], []
            cc_toggle = bool(constitutive_correction)
            if constitutive_correction:
                timepoint_data_ccorr, timepoint_means_extremes_ccorr, regional_timepoint_means_extremes_ccorr = [], [], []
                num_control_df, denom_control_df = get_reference_dfs(num_fluor, denom_fluor)
                df_ccorr_data = initialize_df()
                df_all_times_ccorr = initialize_df()
                df_all_times_error_ccorr = initialize_df()
            df_all_times = initialize_df()
            df_all_times_error = initialize_df()
            
            # begins looping over time points. filters working list of
            # files down to those matching the current time point
            for time_tuple in time_tuples:
                time = time_tuple[0]
                time_hours = time_tuple[1]
                time_flist = [item for item in condition_flist if item[6] == time]
                if not time_flist:
                    continue
                print(f'Processing: {celltype} {condition}, {time}')
                
                # begins looping over (experimental) replicates. filters
                # working list of files down to those matching the
                # current replicate
                df_all_replicates = initialize_df()
                for replicate in replicates:
                    replicate_flist = [item for item in time_flist if (item[3] == replicate) and (item[2] == num_fluor)]
                    if not replicate_flist:
                        continue
                        
                    # generates list of paired files and runs
                    # process_wells to extract data into DataFrames
                    wells = get_well_list(replicate_flist)
                    num_intensities, denom_intensities = process_wells_ratio(wells)
                    
                    if num_intensities.empty or denom_intensities.empty:
                        print(f'Unable to extract usable data from {time} {replicate}.')
                        continue
                        
                    # generates average profile for the current
                    # combination of celltype, condition, time point,
                    # and replicate (i.e., mean of technical replicates)
                    # for each channel
                    avg_num_profile = resample_df(num_intensities).mean(axis = 1)
                    avg_denom_profile = resample_df(denom_intensities).mean(axis = 1)
                    
                    # if there is a (substantial) mistmatch in the
                    # length of the average profile for the numerator
                    # channel and that of the denominator channel (as
                    # might occur in the case of a hypoxia biosensor,
                    # for example), fills missing space with zeroes to
                    # avoid doing math with NaNs
                    if len(avg_num_profile) < len(avg_denom_profile) * 0.8:
                        rep_profile_lengths = [len(denom_intensities[x].dropna().tolist()) for x in denom_intensities.columns]
                        shortfall = len(avg_denom_profile) - len(avg_num_profile)
                        zero_series = pd.Series([0] * shortfall)
                        avg_num_profile = pd.concat([avg_num_profile, zero_series], ignore_index = True)
                    else:
                        num_lengths = [len(num_intensities[x].dropna().tolist()) for x in num_intensities.columns]
                        denom_lengths = [len(denom_intensities[x].dropna().tolist()) for x in denom_intensities.columns]
                        rep_profile_lengths = [x for x in np.mean([num_lengths, denom_lengths], axis = 0, dtype = int)]
                        
                    # reports profile lengths for current replicate so
                    # user can monitor data extraction progress
                    print(f'Profile lengths for {replicate} (radius, {distance_unit}): {[int(x * dist_per_px) for x in rep_profile_lengths]}')
                    avg_profile_length = np.mean(rep_profile_lengths, dtype = "int")
                    
                    # generates average profile for each channel, then
                    # generates average ratio profile. clips lower value
                    # at lowest non-zero value so that the full size of
                    # the specimen is always represented in the graph (as
                    # opposed to the curve running at y=0)
                    avg_denom_profile = [x for x in avg_denom_profile if x > 0]
                    avg_num_profile = pd.Series(resample_list_to(avg_num_profile, avg_profile_length))
                    avg_denom_profile = pd.Series(resample_list_to(avg_denom_profile, avg_profile_length))
                    avg_ratio_profile = avg_num_profile / avg_denom_profile
                    min_ratio = min(i for i in avg_ratio_profile if i > 0)
                    avg_ratio_profile = avg_ratio_profile.clip(lower = min_ratio)
                    
                    # collects the average ratio profile for each
                    # replicate (for a given time point) in a DataFrame
                    df_all_replicates[f'{celltype}_{condition}_{time}_{replicate}'] = avg_ratio_profile
                    
                df_all_replicates = cleanup_df(df_all_replicates)
                if df_all_replicates.empty:
                    print(f'Unable to extract usable data from {time}.')
                    continue
                    
                # resample the DataFrame to the average profile length
                # to allow for mean profile to be calculated (for all
                # replicates at a given time point), and to trim/roll/
                # normalize
                time_profile_lengths = [len(df_all_replicates[x].dropna().tolist()) for x in df_all_replicates.columns]
                df_all_replicates = resample_df(df_all_replicates)
                print(f'Average profile length(s) for {time} (radius, {distance_unit}): {[int(x * dist_per_px) for x in time_profile_lengths]} -- {int(len(df_all_replicates.index) * dist_per_px)} overall\n')
                if trim_values:
                    df_all_replicates = trim_df(trim, "centre-out", df_all_replicates)
                if roll_values:
                    df_all_replicates = roll_df(roll, df_all_replicates)
                if replicate_normalization:
                    df_all_replicates = df_all_replicates / (df_all_replicates.mean() / df_all_replicates.mean().mean())
                    
                # calculates values as required to normalize profile and
                # regional time point data later on, and packages time
                # point data for analysis later (need to know the overall
                # min and max values for overall normalization vs. 1.0 or
                # vs. 100%)
                timepoint_means_extremes.append(df_all_replicates.mean(axis = 1).min())
                timepoint_means_extremes.append(df_all_replicates.mean(axis = 1).max())
                regional_df = get_regional_data(df_all_replicates.dropna())
                regional_timepoint_means_extremes.append(regional_df.mean(axis = 1).min())
                regional_timepoint_means_extremes.append(regional_df.mean(axis = 1).max())
                timepoint_data.append((df_all_replicates, regional_df))
                
                # get mean and error across replicates for a given time
                # point, and aggregate results in "all times" DataFrames
                df_all_times[f'{celltype}_{condition}_{time}'] = df_all_replicates.mean(axis = 1)
                if error_type in ("SE", "SEM"):
                    df_all_times_error[f'{celltype}_{condition}_{time}'] = df_all_replicates.sem(axis = 1)
                elif error_type in ("SD", "STD"):
                    df_all_times_error[f'{celltype}_{condition}_{time}'] = df_all_replicates.std(axis = 1)
                    
                # if constitutive correction is enabled, generate
                # corrected data and do all of the above for that
                # data as well
                if constitutive_correction and cc_toggle:
                    reference_data = get_time_reference(num_control_df, denom_control_df, time_hours)
                    df_ccorr_data[f'{celltype}_{condition}_{time}'] = reference_data
                    df_all_replicates_ccorr = (df_all_replicates.T / pd.Series(resample_list_to(reference_data, len(df_all_replicates)))).T
                    timepoint_means_extremes_ccorr.append(df_all_replicates_ccorr.mean(axis = 1).min())
                    timepoint_means_extremes_ccorr.append(df_all_replicates_ccorr.mean(axis = 1).max())
                    regional_df_ccorr = get_regional_data(df_all_replicates_ccorr.dropna())
                    regional_timepoint_means_extremes_ccorr.append(regional_df_ccorr.mean(axis = 1).min())
                    regional_timepoint_means_extremes_ccorr.append(regional_df_ccorr.mean(axis = 1).max())
                    timepoint_data_ccorr.append((df_all_replicates_ccorr, regional_df_ccorr))
                    
                    df_all_times_ccorr[f'{celltype}_{condition}_{time}'] = df_all_replicates_ccorr.mean(axis = 1)
                    if error_type in ("SE", "SEM"):
                        df_all_times_error_ccorr[f'{celltype}_{condition}_{time}'] = df_all_replicates_ccorr.sem(axis = 1)
                    elif error_type in ("SD", "STD"):
                        df_all_times_error_ccorr[f'{celltype}_{condition}_{time}'] = df_all_replicates_ccorr.std(axis = 1)
                        
            # once all time points are complete, output the data for: all
            # times, individual time points, any graphical options, and
            # an analysis summary
            print(f'Generating outputs for {celltype} {condition}...\n')
            try_dir(dir_output)
            dir_celltype_condition = Path(dir_output, f'{celltype} {condition}')
            try_dir(dir_celltype_condition)
            summary_ratio()
            write_timepoint_ratio_data(timepoint_data, timepoint_means_extremes, regional_timepoint_means_extremes, "no correction")
            df_all_times, df_all_times_error = cleanup_df(df_all_times, df_all_times_error)
            if df_all_times_error.empty:
                df_all_times_error = df_all_times * 0
            write_all_times_ratio_data((df_all_times, df_all_times_error), "no correction")
            info_for_rmaps = [celltype, condition, df_all_times, biosensor, rmaps_cmap, num_fluor, denom_fluor]
            
            # if constitutive correction is enabled, write all the
            # associated data
            if constitutive_correction and cc_toggle:
                write_timepoint_ratio_data(timepoint_data_ccorr, timepoint_means_extremes_ccorr, regional_timepoint_means_extremes_ccorr, "with correction")
                df_all_times_ccorr, df_all_times_error_ccorr = cleanup_df(df_all_times_ccorr, df_all_times_error_ccorr)
                info_for_rmaps.append(df_all_times_ccorr)
                if df_all_times_error_ccorr.empty:
                    df_all_times_error_ccorr = df_all_times_ccorr * 0
                write_all_times_ratio_data((df_all_times_ccorr, df_all_times_error_ccorr), "with correction")
                
                df_ccorr_data = cleanup_df(df_ccorr_data)
                info_for_rmaps.append(df_ccorr_data)
                write_ccorr_data(df_ccorr_data)
            info_for_rmaps.append(cc_toggle)
            
            rmaps_reference_info.append(info_for_rmaps)
            
            # creates any/all types of ratio maps that are indicated by
            # user inputs
            if rmaps_per_time:
                make_rmaps_per_time()
            if rmaps_compare_times:
                make_rmaps_compare_times()
        if rmaps_compare_conditions:
            make_rmaps_compare_conditions()
    if rmaps_compare_celltypes:
        if len(celltypes) > 1:
            rmaps_cmap = default_rmaps_cmap
        make_rmaps_compare_celltypes()

    if not analyzed_celltypes:
        print(f'\nNone of the files in the input directory were suitable for SPoRTS ratio analysis.\nPlease check file naming or switch to the intensity module.')
        
    # output cell below updates user on analysis progression.
    # specifically, checks for unpaired files, reports biosensor and 
    # analysis settings (allows user to cancel if any issues are
    # identified. if not, begins processing based on alphabetical
    # order of celltype --> condition, at the earliest time point.
    # reports profile lengths while analysis is underway
```
